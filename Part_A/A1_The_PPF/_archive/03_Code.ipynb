## Animation 0 | Intro Sequence

```python
%%manim animation_0

class animation_0(Scene):

    def construct(self):
        
        """ Definitions """
        
        colors = sns.color_palette("Blues", 50).as_hex()

        size = 1/6
        n_width = 2
        n_height = 3

        n_rows = len(range(-n_height,n_height+1))
        n_cols = len(range(-n_width,n_width+1))
        w_list = list(range(-n_width,n_width+1))*n_rows
        h_list = [i for i in range(-n_height,n_height+1) for x in 'a'*n_cols]
        block = list(zip(w_list,h_list)) # height: 7, width: 5
        
        string = 'MICROECONOMICS'
        letters = [raster_font[l] for l in string]
        
        """ Run """
                
        shift = 0
        centering = -39
        squares = []
        for l in letters:
            s = [Square(side_length=size, color=config.background_color).move_to(RIGHT*(w + shift*6 + centering)*size + DOWN*h*size) for w,h in [block[i] for i in l]]
            squares = squares + s
            shift = shift + 1
        
        Squares = VGroup(*squares)
        
        self.add(Squares)
        
        for i in range(15):
            update_squares = [s.animate.set_fill(random.sample(colors,1),opacity=1) for s in squares]
            self.play(*update_squares, run_time=1/10)
            self.wait(4/10)
            
        part_label = Tex('{{Part A}} $|$ Episode 1').set_color(GREY).set_color_by_tex_to_color_map(
            {"Part A": BLUE,}
        ).scale(3).move_to(DOWN)
        self.play(AddTextWordByWord(part_label), Squares.animate.to_edge(UP, buff=1))
        
        for i in range(15):
            update_squares = [s.animate.set_fill(random.sample(colors,1),opacity=1) for s in squares]
            self.play(*update_squares, run_time=1/10)
            self.wait(4/10)
        
        self.wait()
```

## Animation 2 | Opportunity Cost

Opportunity cost is the value of the next best alternative: what you give up to make a choice.

```python
%%manim animation_2

class animation_2(Scene):

    def construct(self):
        title_text = Tex("{{Opportunity Cost}} is the value of the next best alternative.").to_edge(UP, buff=1).set_color_by_tex_to_color_map({
            "Opportunity Cost": DEFINITION
        })
        self.play(AddTextWordByWord(title_text))
        
        text_list = ['A','    or    ','B']
        OR = Tex(' or ').scale(1.5)
        #self.add(OR)
        
        A = Tex('A').scale(1.5).next_to(OR,LEFT,buff=2)
        self.play(FadeIn(A))
        
        self.wait(1/2)
        
        self.play(FadeIn(OR))
        
        self.wait(1/2)
        
        B = Tex('B').scale(1.5).next_to(OR,RIGHT,buff=2)
        self.play(FadeIn(B))
        
        self.wait(1/2)
        
        framebox1 = SurroundingRectangle(A, buff = 0.3).set_color(GREEN)
        self.play(Create(framebox1))
        
        self.wait(1/2)
        
        framebox2 = SurroundingRectangle(B, buff = 0.3).set_color(RED)
        self.play(Create(framebox2))
        
        self.wait(1/2)
        
        cost = Tex("Opportunity Cost({{A}}) = {{B}}").scale(1.5).next_to(OR,DOWN,buff=2).set_color_by_tex_to_color_map({
                "A": GREEN,
                "B": RED,
            })
        self.play(Create(cost))
        
        self.wait(1/2)
        
        cost2 = Tex("Opportunity Cost({{B}}) = {{A}}").scale(1.5).next_to(OR,DOWN,buff=2).set_color_by_tex_to_color_map({
                "A": RED,
                "B": GREEN,
            })
        self.play(framebox1.animate.move_to(B),framebox2.animate.move_to(A),Transform(cost,cost2))
        
        self.wait(1/2)
        
        self.play(
            FadeOut(title_text), FadeOut(A), FadeOut(OR), FadeOut(B), 
            FadeOut(framebox1), FadeOut(framebox2),
            FadeOut(cost), FadeOut(cost2),
        )
        self.wait(1/2)
```

## Animation 3 | Molly's Farm

Molly can grow spinach and carrots. She faces opportunity cost of growing both goods. 

```python
%%manim animation_3

class animation_3(MovingCameraScene):

    def construct(self):
        farm_location = LEFT*4 + DOWN/2
        
        farm = Rectangle(height=5, width=4, color=WHITE).move_to(farm_location)
        farm.z_index = 2
        farm_name = Tex("Molly's Farm").scale(1.5).next_to(farm,UP)
        self.play(FadeIn(farm),FadeIn(farm_name))
        
        self.wait(1/2)
        
        spinach = Rectangle(height=5, width=4, color=GREEN, fill_opacity=1).move_to(farm_location)
        carrots = Rectangle(height=0, width=4, color=ORANGE, fill_opacity=1).next_to(spinach,UP,buff=0)
        self.play(FadeIn(carrots),FadeIn(spinach))
        
        spinach_harvest = Tex('{{45}} {{S}}').scale(1.5).move_to(farm_location).set_color_by_tex_to_color_map({
                "S": GREEN,
            })
        self.play(spinach_harvest.animate.move_to(RIGHT*4))
        
        self.wait(1/2)
        
        carrots_new = Rectangle(height=5, width=4, color=ORANGE, fill_opacity=1).move_to(farm_location)
        spinach_new = Rectangle(height=0, width=4, color=GREEN, fill_opacity=1).next_to(carrots_new,DOWN,buff=0)
        self.play(Transform(spinach,spinach_new),Transform(carrots,carrots_new))
        
        self.wait(1/2)
        
        OR = Tex("or").next_to(spinach_harvest,LEFT,buff=1).scale(1.5)
        equal = Tex("=").next_to(spinach_harvest,LEFT,buff=1).scale(1.5)
        
        carrot_harvest = Tex('{{5}} {{C}}').scale(1.5).move_to(LEFT*4).set_color_by_tex_to_color_map({
                "C": ORANGE,
            })
        self.play(FadeIn(OR),carrot_harvest.animate.next_to(OR,LEFT,buff=1))
        
        self.wait(1/2)
        
        self.play(Transform(OR,equal))
        
        self.wait(1/2)
        
        self.play(self.camera.frame.animate.move_to(OR).set(width=OR.width*15))
        
        self.wait(1/2)
        
        spinach_harvest_new = Tex(r"{{$\frac{45}{5}$}} {{S}}").scale(1.5).next_to(OR,RIGHT,buff=1).set_color_by_tex_to_color_map({
                "S": GREEN,
            })
        carrot_harvest_new = Tex(r"$\frac{5}{5}$ {{C}}").scale(1.5).next_to(OR,LEFT,buff=1).set_color_by_tex_to_color_map({
                "C": ORANGE,
            })
        self.play(Transform(carrot_harvest[0],carrot_harvest_new[0]),Transform(spinach_harvest[0],spinach_harvest_new[0]))
        
        self.wait(1/2)
        
        spinach_harvest_new = Tex('{{9}} {{S}}').scale(1.5).next_to(OR,RIGHT,buff=1).set_color_by_tex_to_color_map({
                "S": GREEN,
            })
        carrot_harvest_new = Tex('{{1}} {{C}}').scale(1.5).next_to(OR,LEFT,buff=1).set_color_by_tex_to_color_map({
                "C": ORANGE,
            })
        self.play(Transform(carrot_harvest[0],carrot_harvest_new[0]),Transform(spinach_harvest[0],spinach_harvest_new[0]))
        self.remove(farm_name,farm,spinach,carrots)
        
        self.wait(1/2)
        
        op_cost = Tex("Opportunity Cost(1 {{$C$}}) = 9 {{$S$}}").scale(1.5).next_to(OR,UP,buff=2).set_color_by_tex_to_color_map({
                "$C$": ORANGE,
                "$S$": GREEN,
            })
        self.play(self.camera.frame.animate.move_to(op_cost).set(width=op_cost.width*1.5))
        
        self.wait(1/2)
        
        self.play(AddTextWordByWord(op_cost))
        
        self.wait(1/2)
        
```

## Animation 4 | Molly's Production Possibility Frontier

Show scarcity systematically with a PPF.

```python
%%manim animation_4

class animation_4(MovingCameraScene):

    def construct(self):
        
        PPF_axis_small = Axes(            
            x_range=[0, 55, 5],
            x_length = 7,
            axis_config={"color": WHITE},
            x_axis_config={
                "numbers_to_include": np.arange(0, 60, 15),
                "numbers_with_elongated_ticks": np.arange(0, 60, 10),
                "decimal_number_config": {
                    "num_decimal_places":0,
                    "color":GREEN,
                },
            },
            y_range=[0, 6, 1],
            y_length = 6,
            y_axis_config={
                "numbers_to_include": np.arange(0, 7, 1),
                "numbers_with_elongated_ticks": np.arange(0, 7, 1),
                "decimal_number_config": {
                    "num_decimal_places":0,
                    "color":ORANGE,
                }
            },
            tips=False,
        )
        
        axes = PPF_axis_small.shift(RIGHT*2 + DOWN/2).scale(0.8)
        
        farm_location = LEFT*4 + DOWN/2
        
        farm = Rectangle(height=5, width=4, color=WHITE, fill_opacity=0).move_to(farm_location)
        farm.z_index = 2
        farm_name = Tex("Molly's Farm").scale(1.5).next_to(farm,UP,buff=1/2)
        
        carrots = Rectangle(height=5, width=4, color=ORANGE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        spinach = Rectangle(height=0, width=4, color=GREEN, fill_opacity=1).next_to(carrots,DOWN,buff=0)
        farm_group = VGroup(farm,farm_name,carrots,spinach)
        
        self.add(farm_group)
        self.wait(1/2)
        
        p1 = axes.coords_to_point(0, 5)
        dot = Dot(p1)
        
        t1,s,c,t2 = label = VGroup(
            Text("(").scale(1.2),
            DecimalNumber(
                spinach.get_height()*9,
                show_ellipsis=False,
                num_decimal_places=1,
                include_sign=False,
                color=GREEN
            ).scale(1.2),
            DecimalNumber(
                carrots.get_height(),
                show_ellipsis=False,
                num_decimal_places=1,
                include_sign=False,
                color=ORANGE
            ).scale(1.2),
            Text(")").scale(1.2),
        )
        label.arrange(RIGHT)
        always(label.next_to, dot, UP+RIGHT)
        
        self.play(FadeIn(axes),FadeIn(label),FadeIn(dot))
        self.wait(1/2)
        
        c.add_updater(lambda m: m.set_value(carrots.get_height()).next_to(s))
        s.add_updater(lambda m: m.set_value(spinach.get_height()*9).next_to(t1))
        t2.add_updater(lambda m: m.next_to(c))
        
        d_list = []
        for c_alpha in [1, 0, 1/2, 7/10, 3/10]:
            carrots_new = Rectangle(height=c_alpha*5, width=4, color=ORANGE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
            spinach_new = Rectangle(height=(1-c_alpha)*5, width=4, color=GREEN, fill_opacity=1).next_to(carrots_new,DOWN,buff=0)
            p = axes.coords_to_point((1-c_alpha)*5*9, c_alpha*5)
            self.play(dot.animate.move_to(p),Transform(spinach,spinach_new),Transform(carrots,carrots_new))
            d_list.append(Dot(p, color=WHITE))
            self.add(d_list[-1])
            self.wait(1/2)
        d_group = VGroup()
        d_group.add(*d_list)
            
        def ppf_func(s):
            c = 5 - s/9
            return c
            
        ppf_graph = axes.plot(lambda x: ppf_func(x), color=BLUE, x_range=(0, 45), z_index=-1)
        ppf_group = VGroup(farm,axes)
        ppf_label = Tex("Molly's Production Possibility Frontier").scale(1.5).next_to(ppf_group,UP,buff=1/2)
        ppf_group = VGroup(farm,axes,ppf_label)
        grow = Rectangle(height=5, width=4, color=BLUE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        
        self.play(
            FadeIn(farm),FadeIn(ppf_label),Create(ppf_graph),
            Transform(farm_name, ppf_label), self.camera.frame.animate.move_to(ppf_group),
            FadeIn(grow),
            FadeOut(d_group),FadeOut(dot),FadeOut(t1),FadeOut(s),FadeOut(c),FadeOut(t2)
        )
        self.wait(1/2)
        self.play(FadeOut(farm, ppf_label, ppf_graph, axes, grow, farm_group))
        self.wait(1/4)
        
```

```python
%%manim animation_5

class animation_5(MovingCameraScene):

    def construct(self):
        
        """ Definitions """
        
        PPF_axis = Axes(            
            x_range=[0, 55, 5],
            x_length = 7,
            axis_config={"color": WHITE},
            x_axis_config={
                "numbers_to_include": np.arange(0, 60, 15),
                "numbers_with_elongated_ticks": np.arange(0, 60, 10),
                "decimal_number_config": {
                    "num_decimal_places":0,
                    "color":GREEN,
                },
            },
            y_range=[0, 10, 1],
            y_length = 10,
            y_axis_config={
                "numbers_to_include": np.arange(0, 11, 2),
                "numbers_with_elongated_ticks": np.arange(0, 11, 5),
                "decimal_number_config": {
                    "num_decimal_places":0,
                    "color":ORANGE,
                }
            },
            tips=False,
        )
        
        alpha = 4/5

        PPF_axis_small = Axes(            
            x_range=[0, 55, 5],
            x_length = 7,
            axis_config={"color": WHITE},
            x_axis_config={
                "numbers_to_include": np.arange(0, 60, 15),
                "numbers_with_elongated_ticks": np.arange(0, 60, 10),
                "decimal_number_config": {
                    "num_decimal_places":0,
                    "color":GREEN,
                },
            },
            y_range=[0, 6, 1],
            y_length = 6,
            y_axis_config={
                "numbers_to_include": np.arange(0, 7, 1),
                "numbers_with_elongated_ticks": np.arange(0, 7, 1),
                "decimal_number_config": {
                    "num_decimal_places":0,
                    "color":ORANGE,
                }
            },
            tips=False,
        )
        
        axes = PPF_axis.shift(RIGHT*1 + DOWN/2).scale(0.8*6/10)
        
        """ Starting Objects """
        
        #axes = PPF_axis.shift(RIGHT*2 + DOWN/2).scale(0.8)
        
        farm_location = LEFT*4 + DOWN/2
        farm = Rectangle(height=5, width=4, fill_opacity=0).move_to(farm_location)
        farm.z_index = 2

        farm_name = Tex("Molly's Farm").scale(1.5).next_to(farm,UP,buff=1/2)
        
        carrots = Rectangle(height=5, width=4, color=ORANGE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        carrots_up = Rectangle(height=0, width=4, color=ORANGE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        carrots_down = Rectangle(height=5, width=4, color=ORANGE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        
        spinach = Rectangle(height=0, width=4, color=GREEN, fill_opacity=1).next_to(carrots,DOWN,buff=0)
        spinach_up = Rectangle(height=5, width=4, color=GREEN, fill_opacity=1).next_to(carrots_up,DOWN,buff=0)
        spinach_down = Rectangle(height=0, width=4, color=GREEN, fill_opacity=1).next_to(carrots_down,DOWN,buff=0)

        ppf_graph_0 = axes.plot(PPF_Function, color=GREY, x_range=(0, 45))
        ppf_graph = axes.plot(PPF_Function, color=BLUE, x_range=(0, 45))
        ppf_graph.z_index=1
        
        ppf_subgroup = VGroup(farm, ppf_graph)
        ppf_name = Tex("Molly's Production Possibility Frontier").scale(1.5).to_edge(UP,buff=2/3)
        ppf_group = VGroup(farm, ppf_graph, ppf_name)
        
        farm_group = VGroup(farm,ppf_name)
        
        grow = Rectangle(height=5, width=4, color=BLUE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        
        a = axes.coords_to_point(0, 0)
        b = axes.coords_to_point(45, 0)
        c = axes.coords_to_point(0, 5)
        d = axes.coords_to_point(45*alpha, 0)
        e = axes.coords_to_point(0, 5*alpha)
        f = axes.coords_to_point(0, 8)
        g = axes.coords_to_point(27, 2)

        
        """ Setup Scene """
        
        self.play(FadeIn(axes, grow, farm_group, ppf_graph_0, ppf_graph))
        self.wait()
        #self.play(self.camera.frame.animate.move_to(ppf_group))
        
        """ Show Attainable Region """
        
        attainable = Polygon(a,b,c, color=RED, fill_opacity=0.5)
        attainable.z_index = 0
        attainable_pointer = Arrow(
            np.array([1, 1, 0]),
            np.array([-1, -1, 0]),
            max_tip_length_to_length_ratio=1).move_to(attainable,UP*2)
        attainable_pointer.z_index = 1
        attainable_text = Tex('Attainable Points').set_color(RED).next_to(attainable_pointer,UR)
        
        self.play(FadeIn(attainable, attainable_pointer, attainable_text))
        self.wait()
        self.play(FadeOut(attainable),FadeOut(attainable_pointer),FadeOut(attainable_text))
        self.wait()
        
        """ Shift PPF In """
        
        grow_1 = Rectangle(height=5, width=4*alpha, color=BLUE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        carrots_1 = Rectangle(height=5, width=4*alpha, color=ORANGE, fill_opacity=1).next_to(farm_name,DOWN,buff=1/2)
        spinach_1 = Rectangle(height=0, width=4*alpha, color=GREEN, fill_opacity=1).next_to(carrots,DOWN,buff=0)
        ppf_graph_1 = axes.plot(PPF_Function_New, color=BLUE, x_range=(0, 45*alpha))
        
        self.play(Transform(grow,grow_1),Transform(ppf_graph,ppf_graph_1))#,Transform(carrots,carrots_1),Transform(spinach,spinach_1))
        self.wait()
        
        """ Show Attainable Region """
        
        attainable = Polygon(a,d,e, color=RED, fill_opacity=0.5)
        attainable.z_index = 0
        previously_attainable = Polygon(b,c,e,d, color=PINK, fill_opacity=0.5)
        previously_attainable_text = Tex('No Longer Attainable').set_color(PINK).next_to(attainable_text,UP)
        
        self.play(FadeIn(attainable),FadeIn(attainable_text))
        self.wait()
        self.play(FadeIn(previously_attainable))
        self.wait()
        self.play(FadeIn(previously_attainable_text))
        self.wait()
        self.play(FadeOut(attainable),FadeOut(attainable_text),FadeOut(previously_attainable),FadeOut(previously_attainable_text))
        self.wait()
        
        """ Introduce Better Carrot Technology """
        
        ppf_graph_2 = axes.plot(PPF_Function_Tech,color=BLUE,fill_opacity=1,x_range=(0, 45*alpha))
        spinach_tech = Tex('+ Better Spinach Tech').rotate(PI/2).set_color(BLACK).scale(0.9).move_to(grow)

        self.add(ppf_graph_1.set_color(GREY))
        self.play(Transform(ppf_graph,ppf_graph_2), FadeIn(spinach_tech))
        self.wait()
        
        """ Show Attainable Region """
        
        newly_attainable_text = Tex('Newly Attainable').set_color(YELLOW).next_to(previously_attainable_text,UP)
        
        attainable = Polygon(a,f,d, color=RED, fill_opacity=0.5)
        self.play(FadeIn(attainable))
        self.wait()
        self.play(FadeIn(attainable_text))
        self.wait()
        
        unattainable = Polygon(d,b,g, color=PINK, fill_opacity=0.5)
        self.play(FadeIn(unattainable))
        self.wait()
        self.play(FadeIn(previously_attainable_text))
        self.wait()
        
        newly = Polygon(c,g,f, color=YELLOW, fill_opacity=0.5)
        self.play(FadeIn(newly))
        self.play(FadeIn(newly_attainable_text))
        self.wait()
        
        extra_credit = Polygon(c,e,d,g, color=GREEN, fill_opacity=0.5)
        self.play(FadeIn(extra_credit))
        self.wait()
```
