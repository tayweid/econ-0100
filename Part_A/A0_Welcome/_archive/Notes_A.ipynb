## Tradeoffs


To get a better handle on choices, lets be a bit more sophisticated how we define tradeoffs. Lets say I give you the choice of $A$ OR $B$. If you choose $A$, you can't have $B$. If you choose $B$, you can't have $A$. Economists define opportunity cost as the value of the next best alternative. This definition gives us handy conceptual traction here. If you choose $A$, your next best alternative is $B$. So your opportunity cost of $A$ is $B$. And your opportunity cost of $B$ is $A$. ## Molly's Opportunity Cost

Let's build up some intuition for opportunity cost with a little example. Imagine Molly, a small scale organic farmer, is good at growing spinach and carrots on her $100$ hectare farm. If she spends all her time growing spinach she can grow $45$ tons per year. If instead she spends all her time growing carrots, she can grow $5$ tons per year. If Molly were to switch from growing all carrots to all spinach, she would give up $5$ tons of carrots and instead get $45$ tons of spinach. So we’ll set those two values equal and simplify. And that's it. Molly’s opportunity cost of $1$ unit of carrots is $9$ units of spinach. For every $1$ unit of carrots she wants to grow, she’ll have to give up $9$ units of spinach. This idea is crucial to what we'll do later.


## Molly's Production Possibility Frontier

There is a more systematic way to think about Molly’s farm. We've gone through and calculated Molly's opportunity cost for her farm. But Molly doesn't need to not choose either carrots OR spinach; she can choose any combination of the two crops. So let's go through and examine a range of crop choices she can make. Let's use an $x,y$ graph to plot how much of each crop model is harvesting as we move through a few crop choices. Let's use the vertical axis to represent her carrot harvest and the horizontal axis to represent her spinach harvest. We've already established that she can harvest $5$ tons of carrots and $0$ tons of spinach, or $0$ tons of carrots and $45$ tons of spinach. 


Based on her productivity numbers, if Molly plants carrots on half her land and spinach on the other half, she would have half of $5$ tons of carrots and half of $45$ tons of spinach. Let's also plot that on our $x,y$ graph.


At this point you might wonder how Molly would choose how much of each to plant. Great question. We'll come back to it in later chapters. For now we're just gonna find what's possible to set up the trade-offs involved in our later models.


We can pick another point, splitting the land into $\frac{3}{5}$ carrots and $\frac{2}{5}$ spinach.


As we running through more points its easy to see that any carrot and spinach combination Molly chooses lies on a line which we call the Production Possibility Frontier (PPF). Take a look at the range of other combinations. And guess what? Calculating the the slope of this relationship gives us her opportunity cost. ## Comparative Advantage

Having a lower opportunity cost is what we call **comparative advantage**. And here, Andrew has a comparative advantage in carrots, and Molly has a comparative advantage in spinach. As you might have seen before, there is an important mathematical realization here. When one farmer’s opportunity cost is lower in one crop, the other farmer’s opportunity cost will always be lower in the other.


So Molly realizes that maybe there’s something to this idea, that maybe the guild should be minimizing their joint opportunity cost. So she and Andrew decide to specialize in what each farmer is relatively most efficient at growing, what they have a comparative advantage in. So Molly grows only spinach, growing $45$. And Andrew grows only carrots, growing $4$.


And we can check our work here by checking what happens to our production when we specialize in the good we don't have the comparative advantage in. And our production point is inside the guild's equal-split PPF, so we've correctly chosen the specialization according to comparative advantage. 


So without changing any technology, simply by choosing their crop according to their comparative advantage, the local farmers guild was able to exceed the frontier.
