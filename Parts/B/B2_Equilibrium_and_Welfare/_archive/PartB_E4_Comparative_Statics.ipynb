```python
%%manim animation_

class animation_(Scene):

    def construct(self):
        
        """ Definitions """
        
        colors = sns.color_palette("Blues", 50).as_hex()

        size = 1/6
        n_width = 2
        n_height = 3

        n_rows = len(range(-n_height,n_height+1))
        n_cols = len(range(-n_width,n_width+1))
        w_list = list(range(-n_width,n_width+1))*n_rows
        h_list = [i for i in range(-n_height,n_height+1) for x in 'a'*n_cols]
        block = list(zip(w_list,h_list)) # height: 7, width: 5
        
        string = 'MICROECONOMICS'
        letters = [raster_font[l] for l in string]
        
        """ Run """
                
        shift = 0
        centering = -39
        squares = []
        for l in letters:
            s = [Square(side_length=size, color=config.background_color).move_to(RIGHT*(w + shift*6 + centering)*size + DOWN*h*size) for w,h in [block[i] for i in l]]
            squares = squares + s
            shift = shift + 1
        
        Squares = VGroup(*squares)
        
        self.add(Squares)
        
        for i in range(8):
            update_squares = [s.animate.set_fill(random.sample(colors,1),opacity=1) for s in squares]
            self.play(*update_squares, run_time=1/10)
            self.wait(2/10)
            
        part_label = Tex('{{Part B}} $|$ Episode 4').set_color(GREY).set_color_by_tex_to_color_map(
            {"Part B": BLUE,}
        ).scale(3).next_to(Squares, DOWN*4)
        group = VGroup(Squares, part_label)
        self.play(FadeIn(part_label), group.animate.move_to(0))
        
        for i in range(8):
            update_squares = [s.animate.set_fill(random.sample(colors,1),opacity=1) for s in squares]
            self.play(*update_squares, run_time=1/10)
            self.wait(2/10)
```

```python
""" Axis Pamameters """

x_int = 10
x_max = 50
y_int = 2
y_max = 14

PQ_large = Axes(            
    x_range=[0, x_max, x_int],
    x_length = 7,
    axis_config={"color": WHITE},
    x_axis_config={
        "numbers_to_include": np.arange(0, x_max+x_int, x_int),
        "decimal_number_config": {
            "num_decimal_places":0,
        },
    },
    y_range=[0, y_max, y_int],
    y_length = 6,
    y_axis_config={
        "numbers_to_include": np.arange(0,y_max+y_int,y_int),
        "decimal_number_config": {
            "num_decimal_places":0,
        }
    },
    tips=False,
)

""" Starting Objects """

title_string = "Where do prices come from?"
title = Tex(title_string).to_edge(UP)

supply_axes = PQ_large.copy().scale(0.6).to_edge(DOWN + LEFT).shift(2.5 * UP / 3 + 5 * RIGHT / 4)
supply_y_label = supply_axes.get_y_axis_label("P")
supply_x_label = supply_axes.get_x_axis_label("Q")
supply_group = [supply_axes, supply_y_label, supply_x_label]

demand_axes = PQ_large.copy().scale(0.6).to_edge(DOWN + RIGHT).shift(2.5 * UP / 3 + 5 * LEFT / 4)
demand_y_label = demand_axes.get_y_axis_label("P")
demand_x_label = demand_axes.get_x_axis_label("Q")
demand_group = [demand_axes, demand_y_label, demand_x_label]

""" Functions """

supply_slope = 5
supply_intercept = 2

def Supply(q):
    return supply_intercept + q / supply_slope

def Inv_Supply(p):
    return (p - supply_intercept) * supply_slope

demand_slope = 5
demand_intercept = 12

def Demand(q):
    return demand_intercept - q / demand_slope

def Inv_Demand(p):
    return (demand_intercept - p) * demand_slope

""" Price """

price = ValueTracker()

""" Supply Curve """

Supply_Line = supply_axes.plot(Supply, x_range=[0, x_max]).set_color(YELLOW)
Supply_Line.z_index = 3

""" Supply Equation """

def supply_title_def():
    return MathTex(r"S: P =" + f"{price.get_value():.0f}" + r"= 2 + \frac{Q_s}{5}").next_to(supply_axes, UP * 2).set_color(YELLOW)
supply_title = supply_title_def()

""" Quantity Supplied """

def dot_s_def(p):
    q = Inv_Supply(p)
    dot = Dot(supply_axes.c2p(q, p)).set_color(RED)
    dot.z_index = 3
    return dot
dot_s = dot_s_def(price.get_value())

def p_line_s_def(p):
    return DashedVMobject(supply_axes.plot(lambda x: p, x_range=[-1, x_max])).set_color(RED)
p_line_s = p_line_s_def(price.get_value())

def p_number_s_def(p):
    return DecimalNumber(num_decimal_places=0).set_value(p).set_color(RED).scale(0.8).next_to(p_line_s, LEFT, buff=1/3)
p_number_s = p_number_s_def(price.get_value())

def p_label_s_def(p):
    if Inv_Demand(p) == Inv_Supply(p):
        return Tex("$P^*$:").set_color(RED).next_to(p_number_s, LEFT, buff=1/4)
    else:
        return Tex("$P$:").set_color(RED).next_to(p_number_s, LEFT, buff=1/4)
p_label_s = p_label_s_def(price.get_value())

def q_line_s_def(p):
    q = Inv_Supply(p)
    return DashedVMobject(Line(supply_axes.c2p(q, 2), supply_axes.c2p(q, p))).set_color(RED)
q_line_s = q_line_s_def(price.get_value())

def q_number_s_def(p):
    q = Inv_Supply(p)
    q_intercept = supply_axes.c2p(q, 0)
    return DecimalNumber(num_decimal_places=0).set_value(q).set_color(RED).scale(0.8).next_to(q_intercept, DOWN, buff=1/3)
q_number_s = q_number_s_def(price.get_value())

def q_label_s_def(p):
    q = Inv_Supply(p)
    q_intercept = supply_axes.c2p(q, 0)
    return Tex("$Q_s$").set_color(RED).next_to(q_intercept, UP, buff=0)
q_label_s = q_label_s_def(price.get_value())

quantity_supplied_group = [dot_s, p_line_s, p_number_s, p_label_s, q_line_s, q_number_s, q_label_s]

""" Demand Line """

Demand_Line = demand_axes.plot(Demand, x_range=[0, x_max]).set_color(BLUE)
Demand_Line.z_index = 3

""" Demand Equation """

def demand_title_def():
    return MathTex(r"D: P =" + f"{price.get_value():.0f}" + r"= 12 - \frac{Q_b}{5}").next_to(demand_axes, UP * 2).set_color(BLUE)
demand_title = demand_title_def()

""" Quantity Demanded """

def dot_d_def(p):
    q = Inv_Demand(p)
    dot = Dot(demand_axes.c2p(q, p)).set_color(RED)
    dot.z_index = 3
    return dot
dot_d = dot_d_def(price.get_value())

def p_line_d_def(p):
    return DashedVMobject(demand_axes.plot(lambda x: p, x_range=[-1, x_max])).set_color(RED)
p_line_d = p_line_d_def(price.get_value())

def p_number_d_def(p):
    return DecimalNumber(num_decimal_places=0).set_value(p).set_color(RED).scale(0.8).next_to(p_line_d, LEFT, buff=1/3)
p_number_d = p_number_d_def(price.get_value())

def p_label_d_def(p):
    if Inv_Demand(p) == Inv_Supply(p):
        return Tex("$P^*$:").set_color(RED).next_to(p_number_d, LEFT, buff=1/4)
    else:
        return Tex("$P$:").set_color(RED).next_to(p_number_d, LEFT, buff=1/4)
p_label_d = p_label_d_def(price.get_value())

def q_line_d_def(p):
    q = Inv_Demand(p)
    return DashedVMobject(Line(demand_axes.c2p(q, 2), demand_axes.c2p(q, p))).set_color(RED)
q_line_d = q_line_d_def(price.get_value())

def q_number_d_def(p):
    q = Inv_Demand(p)
    q_intercept = demand_axes.c2p(q, 0)
    return DecimalNumber(num_decimal_places=0).set_value(q).set_color(RED).scale(0.8).next_to(q_intercept, DOWN, buff=1/3)
q_number_d = q_number_d_def(price.get_value())

def q_label_d_def(p):
    q = Inv_Demand(p)
    q_intercept = demand_axes.c2p(q, 0)
    return Tex("$Q_b$").set_color(RED).next_to(q_intercept, UP, buff=0)
q_label_d = q_label_d_def(price.get_value())

quantity_demanded_group = [dot_d, p_line_d, p_number_d, p_label_d, q_line_d, q_number_d, q_label_d]

""" Exchange """

def shortage_line_def(p):
    qb = Inv_Demand(p)
    qs = Inv_Supply(p)
    if qb > qs:
        shortage_line = Line(demand_axes.c2p(qb, 0), demand_axes.c2p(qs, 0)).set_color(PINK)
    elif qb < qs:
        shortage_line = Line(supply_axes.c2p(qb, 0), supply_axes.c2p(qs, 0)).set_color(PINK)
    else:
        shortage_line = Line(demand_axes.c2p(qb, 0), demand_axes.c2p(qs, 0)).set_color(RED)
    return shortage_line
shortage_line = shortage_line_def(price.get_value())

def shortage_label_def(p):
    qb = Inv_Demand(p)
    qs = Inv_Supply(p)
    if qb > qs:
        shortage_label = Tex('Shortage').scale(0.5).set_color(PINK).next_to(shortage_line, UP, buff=0)
    elif qb < qs:
        shortage_label = Tex('Surplus').scale(0.5).set_color(PINK).next_to(shortage_line, UP, buff=0)
        qx = qb
    else:
        shortage_label = Tex('').scale(0.5).set_color(PINK).next_to(shortage_line, UP, buff=0)
    return shortage_label
shortage_label = shortage_label_def(price.get_value())

def supply_qx_label_def(p):
    qx = min(Inv_Demand(p), Inv_Supply(p))
    supply_qx = supply_axes.c2p(qx, 0)
    if Inv_Demand(p) == Inv_Supply(p):
        return Tex("$Q^*$").set_color(RED).next_to(supply_qx, DOWN, buff=3/4)
    else:
        return Tex("$Q_x$").set_color(RED).next_to(supply_qx, DOWN, buff=3/4)
supply_qx_label = supply_qx_label_def(price.get_value())

def supply_qx_dot_def(p):
    qx = min(Inv_Demand(p), Inv_Supply(p))
    supply_qx = supply_axes.c2p(qx, 0)
    line = Line(supply_axes.c2p(qx, -0.5), supply_axes.c2p(qx, 0.5)).set_color(RED)
    line.z_index = 1
    return line
supply_qx_dot = supply_qx_dot_def(price.get_value())

def demand_qx_label_def(p):
    qx = min(Inv_Demand(p), Inv_Supply(p))
    demand_qx = demand_axes.c2p(qx, 0)
    if Inv_Demand(p) == Inv_Supply(p):
        return Tex("$Q^*$").set_color(RED).next_to(demand_qx, DOWN, buff=3/4)
    else:
        return Tex("$Q_x$").set_color(RED).next_to(demand_qx, DOWN, buff=3/4)
demand_qx_label = demand_qx_label_def(price.get_value())

def demand_qx_dot_def(p):
    qx = min(Inv_Demand(p), Inv_Supply(p))
    supply_qx = demand_axes.c2p(qx, 0)
    line = Line(demand_axes.c2p(qx, -0.5), demand_axes.c2p(qx, 0.5)).set_color(RED)
    line.z_index = 1
    return line
demand_qx_dot = demand_qx_dot_def(price.get_value())

exchange_group = [shortage_line, shortage_label, supply_qx_label, demand_qx_label, demand_qx_dot]

""" Producer Surplus """

def ps_def(p):
    q = min(Inv_Supply(p), Inv_Demand(p))
    lines = []
    
    for i in np.arange(2/3, q, 2/3):
        i_p = Supply(i)
        
        ps_line = Line(supply_axes.c2p(i,p), supply_axes.c2p(i,i_p)).set_color(YELLOW).set_opacity(0.5)
        ps_line.z_index = -1
        lines.append(ps_line)
        
    return VGroup(*lines)
ps = ps_def(price.get_value())

""" Consumer Surplus """

def cs_def(p):
    q = min(Inv_Supply(p), Inv_Demand(p))
    lines = []
    
    for i in np.arange(2/3, q, 2/3):
        i_p = Demand(i)
        
        cs_line = Line(supply_axes.c2p(i,p), supply_axes.c2p(i,i_p)).set_color(BLUE).set_opacity(0.5)
        cs_line.z_index = -1
        lines.append(cs_line)
        
    return VGroup(*lines)
cs = cs_def(price.get_value())

""" Update Helpers """

updater_dict = {
    supply_title: lambda m: m.become(supply_title_def()),
    dot_s: lambda m: m.become(dot_s_def(price.get_value())),
    p_line_s: lambda m: m.become(p_line_s_def(price.get_value())),
    p_number_s: lambda m: m.become(p_number_s_def(price.get_value())),
    p_label_s: lambda m: m.become(p_label_s_def(price.get_value())),
    q_line_s: lambda m: m.become(q_line_s_def(price.get_value())),
    q_number_s: lambda m: m.become(q_number_s_def(price.get_value())),
    q_label_s: lambda m: m.become(q_label_s_def(price.get_value())),
    demand_title: lambda m: m.become(demand_title_def()),
    dot_d: lambda m: m.become(dot_d_def(price.get_value())),
    p_line_d: lambda m: m.become(p_line_d_def(price.get_value())),
    p_number_d: lambda m: m.become(p_number_d_def(price.get_value())),
    p_label_d: lambda m: m.become(p_label_d_def(price.get_value())),
    q_line_d: lambda m: m.become(q_line_d_def(price.get_value())),
    q_number_d: lambda m: m.become(q_number_d_def(price.get_value())),
    q_label_d: lambda m: m.become(q_label_d_def(price.get_value())),
    shortage_line: lambda m: m.become(shortage_line_def(price.get_value())),
    shortage_label: lambda m: m.become(shortage_label_def(price.get_value())),
    supply_qx_label: lambda m: m.become(supply_qx_label_def(price.get_value())),
    supply_qx_dot: lambda m: m.become(supply_qx_dot_def(price.get_value())),
    demand_qx_label: lambda m: m.become(demand_qx_label_def(price.get_value())),
    demand_qx_dot: lambda m: m.become(demand_qx_dot_def(price.get_value())),
    ps: lambda m: m.become(ps_def(price.get_value())),
    cs: lambda m: m.become(cs_def(price.get_value())),
}

def Add_Updater(mobjects: list):
    for m in mobjects:
        try:
            m.add_updater(updater_dict[m])
        except:
            print('pass')

def Remove_Updater(mobjects: list):
    for m in mobjects:
        try:
            m.remove_updater(updater_dict[m])
        except:
            pass

def swait(scene, time=1/2):
    scene.wait(time)
```

```python
%%manim animation_2

class animation_2(MovingCameraScene):

    def construct(self):
        
        """ Starting Objects """
        
        title_string = "How `good' are markets anyway?"
        title = Tex(title_string).to_edge(UP)

        axes = PQ_large.copy().scale(0.7).to_edge(DOWN).shift(0.9*UP+LEFT*2)
        y_label = axes.get_y_axis_label("P")
        x_label = axes.get_x_axis_label("Q")
        grid_labels = VGroup(x_label, y_label)

        """ Functions """
        
        supply_slope = 5
        supply_intercept = 2

        def Supply(q):
            return supply_intercept + q / supply_slope

        def Inv_Supply(p):
            return (p - supply_intercept) * supply_slope

        demand_slope = 5
        demand_intercept = 12

        def Demand(q):
            return demand_intercept - q / demand_slope

        def Inv_Demand(p):
            return (demand_intercept - p) * demand_slope

        """ Price """
        
        price = ValueTracker(4)

        """ Supply and Demand Curves """
        
        Supply_Line = axes.plot(Supply, x_range=[0, x_max]).set_color(YELLOW)
        Supply_Line.z_index = 3
        
        Demand_Line = axes.plot(Demand, x_range=[0, x_max]).set_color(BLUE)
        Demand_Line.z_index = 3

        """ Equations """
        
        def supply_title_def():
            return MathTex(r"S: P = 2 + \frac{Q_s}{5}").next_to(axes, UP).shift(2.5*DOWN).to_edge(RIGHT).set_color(YELLOW)
        supply_title = supply_title_def()

        def demand_title_def():
            return MathTex(r"D: P = 12 - \frac{Q_b}{5}").next_to(axes, UP).shift(4*DOWN).to_edge(RIGHT).set_color(BLUE)
        demand_title = demand_title_def()

        """ Start """
        
        self.play(FadeIn(title, axes, grid_labels, Supply_Line, Demand_Line, supply_title, demand_title))

        """ Quantity Supplied """

        def dot_s_def(p):
            q = Inv_Supply(p)
            dot = Dot(axes.c2p(q, p)).set_color(RED)
            dot.z_index = 3
            return dot
        dot_s = dot_s_def(price.get_value())
        dot_s.add_updater(lambda m: m.become(dot_s_def(price.get_value())))

        #def p_line_s_def(p):
        #    return DashedVMobject(axes.plot(lambda x: p, x_range=[-1, x_max])).set_color(RED)
        #p_line_s = p_line_s_def(price.get_value())
        #p_line_s.add_updater(lambda m: m.become(p_line_s_def(price.get_value())))

        #def p_number_s_def(p):
        #    return Line(axes.c2p(-2.5, p), axes.c2p(2, p), stroke_width=5).set_color(RED)
        #p_number_s = p_number_s_def(price.get_value())
        #p_number_s.add_updater(lambda m: m.become(p_number_s_def(price.get_value())))

        #def p_label_s_def(p):
        #    if Inv_Demand(p) == Inv_Supply(p):
        #        return Tex("$P^*$").set_color(RED).next_to(p_number_s, LEFT, buff=1/3+0.1)
        #    else:
        #        return Tex("$P$").set_color(RED).next_to(p_number_s, LEFT, buff=1/3+0.1)
        #p_label_s = p_label_s_def(price.get_value())
        #p_label_s.add_updater(lambda m: m.become(p_label_s_def(price.get_value())))

        def q_line_s_def(p):
            q = Inv_Supply(p)
            return DashedVMobject(Line(axes.c2p(q, 0), axes.c2p(q, p))).set_color(RED)
        q_line_s = q_line_s_def(price.get_value())
        q_line_s.add_updater(lambda m: m.become(q_line_s_def(price.get_value())))

        def q_number_s_def(p):
            q = Inv_Supply(p)
            q_intercept = axes.c2p(q, 0)
            return DecimalNumber(num_decimal_places=0).set_value(q).set_color(RED).scale(0.8).next_to(q_intercept, DOWN, buff=1/3+0.1)
        q_number_s = q_number_s_def(price.get_value())
        q_number_s.add_updater(lambda m: m.become(q_number_s_def(price.get_value())))

        def q_dot_s_def(p):
            q = Inv_Supply(p)
            line = Line(axes.c2p(q, -0.5), axes.c2p(q, 0.5)).set_color(RED)
            line.z_index = 1
            return line
        q_dot_s = q_dot_s_def(price.get_value())
        q_dot_s.add_updater(lambda m: m.become(q_dot_s_def(price.get_value())))

        def q_label_s_def(p):
            q = Inv_Supply(p)
            q_intercept = axes.c2p(q, 0)
            if Inv_Demand(p) == Inv_Supply(p):
                return Tex("$Q^*$").set_color(RED).next_to(q_intercept, DOWN, buff=3/4)
            else:
                return Tex("$Q_s$").set_color(RED).next_to(q_intercept, DOWN, buff=3/4)
        q_label_s = q_label_s_def(price.get_value())
        q_label_s.add_updater(lambda m: m.become(q_label_s_def(price.get_value())))

        """ Quantity Demanded """

        def dot_d_def(p):
            q = Inv_Demand(p)
            dot = Dot(axes.c2p(q, p)).set_color(RED)
            dot.z_index = 3
            return dot
        dot_d = dot_d_def(price.get_value())
        dot_d.add_updater(lambda m: m.become(dot_d_def(price.get_value())))

        def p_line_d_def(p):
            return DashedVMobject(axes.plot(lambda x: p, x_range=[-1, x_max])).set_color(RED)
        p_line_d = p_line_d_def(price.get_value())
        p_line_d.add_updater(lambda m: m.become(p_line_d_def(price.get_value())))

        def p_number_d_def(p):
            return Line(axes.c2p(-2.5, p), axes.c2p(2, p), stroke_width=5).set_color(RED)
        p_number_d = p_number_d_def(price.get_value())
        p_number_d.add_updater(lambda m: m.become(p_number_d_def(price.get_value())))

        def p_label_d_def(p):
            if Inv_Demand(p) == Inv_Supply(p):
                return Tex("$P^*$").set_color(RED).next_to(p_number_d, LEFT, buff=1/3+0.1)
            else:
                return Tex("$P$").set_color(RED).next_to(p_number_d, LEFT, buff=1/3+0.1)
        p_label_d = p_label_d_def(price.get_value())
        p_label_d.add_updater(lambda m: m.become(p_label_d_def(price.get_value())))

        def q_line_d_def(p):
            q = Inv_Demand(p)
            return DashedVMobject(Line(axes.c2p(q, 0), axes.c2p(q, p))).set_color(RED)
        q_line_d = q_line_d_def(price.get_value())
        def q_line_d_update(m):
            m.become(q_line_d_def(price.get_value()))
        q_line_d.add_updater(q_line_d_update)

        def q_number_d_def(p):
            q = Inv_Demand(p)
            q_intercept = axes.c2p(q, 0)
            return DecimalNumber(num_decimal_places=0).set_value(q).set_color(RED).scale(0.8).next_to(q_intercept, DOWN, buff=1/3+0.1)
        q_number_d = q_number_d_def(price.get_value())
        q_number_d.add_updater(lambda m: m.become(q_number_d_def(price.get_value())))

        def q_dot_d_def(p):
            q = Inv_Demand(p)
            line = Line(axes.c2p(q, -0.5), axes.c2p(q, 0.5)).set_color(RED)
            line.z_index = 1
            return line
        q_dot_d = q_dot_d_def(price.get_value())
        q_dot_d.add_updater(lambda m: m.become(q_dot_d_def(price.get_value())))

        def q_label_d_def(p):
            q = Inv_Demand(p)
            q_intercept = axes.c2p(q, 0)
            if Inv_Demand(p) == Inv_Supply(p):
                return Tex("$Q^*$").set_color(RED).next_to(q_intercept, DOWN, buff=3/4)
            else:
                return Tex("$Q_b$").set_color(RED).next_to(q_intercept, DOWN, buff=3/4)
        q_label_d = q_label_d_def(price.get_value())
        q_label_d.add_updater(lambda m: m.become(q_label_d_def(price.get_value())))

        """ Exchange """

        def shortage_line_def(p):
            qb = Inv_Demand(p)
            qs = Inv_Supply(p)
            if qb > qs:
                shortage_line = Line(axes.c2p(qb, 0), axes.c2p(qs, 0)).set_color(PINK)
            elif qb < qs:
                shortage_line = Line(axes.c2p(qb, 0), axes.c2p(qs, 0)).set_color(PINK)
            else:
                shortage_line = Line(axes.c2p(qb, 0), axes.c2p(qs, 0)).set_color(RED)
            return shortage_line
        shortage_line = shortage_line_def(price.get_value())
        shortage_line.add_updater(lambda m: m.become(shortage_line_def(price.get_value())))

        def shortage_label_def(p):
            qb = Inv_Demand(p)
            qs = Inv_Supply(p)
            if qb > qs:
                shortage_label = Tex('Shortage').scale(0.6).set_color(PINK).next_to(shortage_line, UP, buff=0)
            elif qb < qs:
                shortage_label = Tex('Surplus').scale(0.6).set_color(PINK).next_to(shortage_line, UP, buff=0)
                qx = qb
            else:
                shortage_label = Tex('').scale(0.6).set_color(PINK).next_to(shortage_line, UP, buff=0)
            return shortage_label
        shortage_label = shortage_label_def(price.get_value())
        shortage_label.add_updater(lambda m: m.become(shortage_label_def(price.get_value())))

        """ Producer Surplus """
        
        def ps_def(p):
            q = min(Inv_Supply(p), Inv_Demand(p))
            lines = []
            
            for i in np.arange(2/3, q, 2/3):
                i_p = Supply(i)
                
                ps_line = Line(axes.c2p(i,p), axes.c2p(i,i_p)).set_color(YELLOW).set_opacity(0.5)
                ps_line.z_index = -1
                lines.append(ps_line)
                
            return VGroup(*lines)
        ps = ps_def(price.get_value())
        ps.add_updater(lambda m: m.become(ps_def(price.get_value())))

        """ Consumer Surplus """
        
        def cs_def(p):
            q = min(Inv_Supply(p), Inv_Demand(p))
            lines = []
            
            for i in np.arange(2/3, q, 2/3):
                i_p = Demand(i)
                
                cs_line = Line(axes.c2p(i,p), axes.c2p(i,i_p)).set_color(BLUE).set_opacity(0.5)
                cs_line.z_index = -1
                lines.append(cs_line)
                
            return VGroup(*lines)
        cs = cs_def(price.get_value())
        cs.add_updater(lambda m: m.become(cs_def(price.get_value())))

        """ DWL """
        
        def dwl_def(p):
            q = min(Inv_Supply(p), Inv_Demand(p))
            lines = []
            
            for i in np.arange(q, 25, 2/3):                
                dwl_line = Line(axes.c2p(i,Supply(i)), axes.c2p(i,Demand(i))).set_color(GREY).set_opacity(0.5)
                dwl_line.z_index = -1
                lines.append(dwl_line)
                
            return VGroup(*lines)
        dwl = dwl_def(price.get_value())
        dwl.add_updater(lambda m: m.become(dwl_def(price.get_value())))

        """ Price Ceiling """

        price_ceiling = ValueTracker(5)

        def price_ceiling_def(p):
            line = Line(axes.c2p(-0.2, 0), axes.c2p(-0.2, p), stroke_width=10).set_color(GREEN)
            line.z_index = -1
            return line
        price_ceiling_line = price_ceiling_def(price_ceiling.get_value())
        price_ceiling_line.add_updater(lambda m: m.become(price_ceiling_def(price_ceiling.get_value())))

        def price_ceiling_dot_def(p):
            line = Line(axes.c2p(-2.5, p), axes.c2p(2, p), stroke_width=5).set_color(GREEN)
            line.z_index = -1
            return line
        price_ceiling_dot = price_ceiling_dot_def(price_ceiling.get_value())
        price_ceiling_dot.add_updater(lambda m: m.become(price_ceiling_dot_def(price_ceiling.get_value())))

        def price_ceiling_label_def():
            return Tex('Legal').scale(0.6).set_color(GREEN).rotate(np.pi/2).next_to(price_ceiling_line, LEFT, buff=0.3)
        price_ceiling_label = price_ceiling_label_def()
        price_ceiling_label.add_updater(lambda m: m.become(price_ceiling_label_def()))

        """ Price Floor """

        price_floor = ValueTracker(10)

        def price_floor_def(p):
            line = Line(axes.c2p(-0.2, p), axes.c2p(-0.2, 14), stroke_width=10).set_color(GREEN)
            line.z_index = -1
            return line
        price_floor_line = price_floor_def(price_floor.get_value())
        price_floor_line.add_updater(lambda m: m.become(price_floor_def(price_floor.get_value())))

        def price_floor_dot_def(p):
            line = Line(axes.c2p(-2.5, p), axes.c2p(2, p), stroke_width=5).set_color(GREEN)
            line.z_index = -1
            return line
        price_floor_dot = price_floor_dot_def(price_floor.get_value())
        price_floor_dot.add_updater(lambda m: m.become(price_floor_dot_def(price_floor.get_value())))

        def price_floor_label_def():
            return Tex('Legal').scale(0.6).set_color(GREEN).rotate(np.pi/2).next_to(price_floor_line, LEFT, buff=0.3)
        price_floor_label = price_floor_label_def()
        price_floor_label.add_updater(lambda m: m.become(price_floor_label_def()))

        """ Price Ceiling """

        new_title_string = "{{Welfare Analysis}}: the study of the benefits of markets."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Welfare Analysis": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
        new_title_string = "{{Price Controls}}: government price restrictions."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Price Controls": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
        new_title_string = "Can the government improve the market with a {{Price Ceiling}}?"
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Price Ceiling": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
        self.play(FadeIn(price_ceiling_line, price_ceiling_label, price_ceiling_dot))
        self.wait()
        self.play(FadeIn(p_label_d, p_line_d, p_number_d))
        self.wait()
        self.play(FadeIn(dot_d, q_line_d, q_number_d, q_dot_d, q_label_d))
        self.wait()
        self.play(FadeIn(dot_s, q_line_s, q_number_s, q_dot_s, q_label_s))
        self.wait()
        self.play(FadeIn(shortage_line, shortage_label))
        self.wait()
        
        new_title_string = "Buyers and sellers push as close as possible to {{Equilibrium}}."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Equilibrium": PURPLE,}).to_edge(UP).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        self.play(price.animate.set_value(5))
        self.wait()
        
        new_title_string = "A binding {{Price Celing}} creates a shortage."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Price Ceiling": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        new_title_string = "How `good' is this market?"
        new_title = Tex(new_title_string).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
        self.play(Create(ps))
        self.wait()
        self.play(Create(cs))
        self.wait()
        self.play(Create(dwl))
        self.wait()

        new_title_string = "A shortage creates {{Deadweight Loss}}."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Deadweight Loss": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        new_title_string = "{{Deadweight Loss}}: a loss of surplus value for society as a whole."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Deadweight Loss": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
        self.play(price.animate.set_value(7), price_ceiling.animate.set_value(7))
        self.wait()
        
        new_title_string = "There is no DWL in {{Competitive Markets}} with no externalities."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Competitive Markets": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        """ Price Floor """

        new_title_string = "Can the government improve the market with a {{Price Floor}}?"
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Price Floor": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        self.play(FadeOut(
                    price_ceiling_line, price_ceiling_label, price_ceiling_dot,
                    p_label_d, p_line_d, p_number_d,
                    dot_d, q_line_d, q_number_d, q_dot_d, q_label_d,
                    dot_s, q_line_s, q_number_s, q_dot_s, q_label_s,
                    shortage_line, shortage_label,
                    ps, cs, dwl
                    ),
                  Transform(title, new_title)
                 )
        self.wait()

        price.set_value(11)
        
        self.play(FadeIn(price_floor_line, price_floor_label, price_floor_dot))
        self.wait()
        self.play(FadeIn(p_label_d, p_line_d, p_number_d))
        self.wait()
        self.play(FadeIn(dot_d, q_line_d, q_number_d, q_dot_d, q_label_d))
        self.wait()
        self.play(FadeIn(dot_s, q_line_s, q_number_s, q_dot_s, q_label_s))
        self.wait()
        self.play(FadeIn(shortage_line, shortage_label))
        self.wait()

        new_title_string = "Buyers and sellers push as close as possible to equilibrium."
        new_title = Tex(new_title_string).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        self.play(price.animate.set_value(10))
        self.wait()
        
        new_title_string = "A binding Price Floor creates a surplus."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Price Floor": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        new_title_string = "How `good' is this market?"
        new_title = Tex(new_title_string).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
        self.play(Create(ps))
        self.wait()
        self.play(Create(cs))
        self.wait()

        new_title_string = "A surplus also creates {{Deadweight Loss}}."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Deadweight Loss": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
        self.play(Create(dwl))
        self.wait()
        
        self.play(price.animate.set_value(7), price_floor.animate.set_value(7))
        self.wait()
        
        new_title_string = "There is no DWL in {{Competitive Markets}} with no externalities."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Competitive Markets": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()
        
```

```python
%%manim animation_4

class animation_4(MovingCameraScene):

    def construct(self):
        
        """ Starting Objects """
        
        title_string = "How do prices change?"
        title = Tex(title_string).to_edge(UP)

        axes = PQ_large.copy().scale(0.7).to_edge(DOWN).shift(0.9*UP+LEFT*2)
        y_label = axes.get_y_axis_label("P")
        x_label = axes.get_x_axis_label("Q")
        grid_labels = VGroup(x_label, y_label)

        """ Price and Parameters """
        
        price = ValueTracker(7)
        l = ValueTracker(0)
        m = ValueTracker(0)

        """ Functions """
        
        supply_slope = 5
        supply_intercept = 2

        def Supply(q):
            return supply_intercept + l.get_value() + q / supply_slope

        def Inv_Supply(p):
            return (p - supply_intercept - l.get_value()) * supply_slope

        demand_slope = 5
        demand_intercept = 12

        def Demand(q):
            return demand_intercept + m.get_value() - q / demand_slope

        def Inv_Demand(p):
            return (demand_intercept + m.get_value() - p) * demand_slope

        """ Equations """
        
        def supply_title_def():
            return MathTex(r"S: P = 2 + z_s + \frac{Q_s}{5}").next_to(axes, UP).shift(2.5*DOWN).to_edge(RIGHT).set_color(YELLOW)
        supply_title = supply_title_def()

        def demand_title_def():
            return MathTex(r"D: P = 12 + z_b - \frac{Q_b}{5}").next_to(axes, UP).shift(4*DOWN).to_edge(RIGHT).set_color(BLUE)
        demand_title = demand_title_def()

        """ Supply and Demand Curves """
        
        Supply_Line = axes.plot(Supply, x_range=[0, x_max]).set_color(YELLOW)
        Supply_Line.z_index = 3
        
        Demand_Line = axes.plot(Demand, x_range=[0, x_max]).set_color(BLUE)
        Demand_Line.z_index = 3

        """ Quantity Demanded """

        def dot_d_def(p):
            q = Inv_Demand(p)
            dot = Dot(axes.c2p(q, p)).set_color(RED)
            dot.z_index = 3
            return dot
        dot_d = dot_d_def(price.get_value())
        dot_d.add_updater(lambda m: m.become(dot_d_def(price.get_value())))

        def p_line_d_def(p):
            return DashedVMobject(axes.plot(lambda x: p, x_range=[-1, x_max])).set_color(RED)
        p_line_d = p_line_d_def(price.get_value())
        p_line_d.add_updater(lambda m: m.become(p_line_d_def(price.get_value())))

        def p_number_d_def(p):
            return Line(axes.c2p(-2.5, p), axes.c2p(2, p), stroke_width=5).set_color(RED)
        p_number_d = p_number_d_def(price.get_value())
        p_number_d.add_updater(lambda m: m.become(p_number_d_def(price.get_value())))

        def p_label_d_def(p):
            if Inv_Demand(p) == Inv_Supply(p):
                return Tex("$P^*$").set_color(RED).next_to(p_number_d, LEFT, buff=1/3+0.1)
            else:
                return Tex("$P$").set_color(RED).next_to(p_number_d, LEFT, buff=1/3+0.1)
        p_label_d = p_label_d_def(price.get_value())
        p_label_d.add_updater(lambda m: m.become(p_label_d_def(price.get_value())))

        def q_line_d_def(p):
            q = Inv_Demand(p)
            return DashedVMobject(Line(axes.c2p(q, 0), axes.c2p(q, p))).set_color(RED)
        q_line_d = q_line_d_def(price.get_value())
        def q_line_d_update(m):
            m.become(q_line_d_def(price.get_value()))
        q_line_d.add_updater(q_line_d_update)

        def q_dot_d_def(p):
            q = Inv_Demand(p)
            line = Line(axes.c2p(q, -0.5), axes.c2p(q, 0.5)).set_color(RED)
            line.z_index = 1
            return line
        q_dot_d = q_dot_d_def(price.get_value())
        q_dot_d.add_updater(lambda m: m.become(q_dot_d_def(price.get_value())))

        def q_label_d_def(p):
            q = Inv_Demand(p)
            q_intercept = axes.c2p(q, 0)
            if Inv_Demand(p) == Inv_Supply(p):
                return Tex("$Q^*$").set_color(RED).next_to(q_intercept, DOWN, buff=1/3+0.1)
            else:
                return Tex("$Q_b$").set_color(RED).next_to(q_intercept, DOWN, buff=3/4)
        q_label_d = q_label_d_def(price.get_value())
        q_label_d.add_updater(lambda m: m.become(q_label_d_def(price.get_value())))

        """ Equilibrium """

        def p_star():
            return supply_intercept + l.get_value() + (demand_intercept - supply_intercept + m.get_value() - l.get_value())/2

        def q_star():
            return 5 * (demand_intercept - supply_intercept + m.get_value() - l.get_value()) / 2

        """ Comparative Statics """
        
        self.play(FadeIn(
            title, axes, grid_labels, Supply_Line, Demand_Line, supply_title, demand_title,
            dot_d, p_line_d, p_number_d, p_label_d, q_line_d, q_dot_d, q_label_d
        ))
        self.wait()

        new_title_string = "{{Comparative Statics}}: how external forces change equilibrium."
        new_title = Tex(new_title_string).set_color_by_tex_to_color_map({"Comparative Statics": PURPLE,}).to_edge(UP)
        self.play(Transform(title, new_title))
        self.wait()

        self.play(FadeOut(
            axes, grid_labels, Supply_Line, Demand_Line, supply_title, demand_title,
            dot_d, p_line_d, p_number_d, p_label_d, q_line_d, q_dot_d, q_label_d
        ),
                  title.animate.move_to(ORIGIN)
                 )
        self.wait()
```

```python
%%manim animation_1

class animation_1(MovingCameraScene):

    def construct(self):
        
        """ Definitions """
        
        PQ_axis = Axes(            
            x_range=[0, 1000, 500],
            x_length = 7,
            axis_config={"color": WHITE},
            x_axis_config={
                "numbers_to_include": np.arange(0, 1000, 500),
                "decimal_number_config": {
                    "num_decimal_places":0,
                },
            },
            y_range=[0, 100, 50],
            y_length = 6,
            y_axis_config={
                "numbers_to_include": [100,50,0],
                "decimal_number_config": {
                    "num_decimal_places":0,
                }
            },
            tips=False,
        )
        
        """ Starting Objects """
        
        axes = PQ_axis.copy().scale(0.8).to_edge(DOWN).shift(UP/2)

        y_label = axes.get_y_axis_label("P")
        x_label = axes.get_x_axis_label("Q")
        grid_labels = VGroup(x_label, y_label)
        
        self.play(FadeIn(axes), FadeIn(grid_labels))
        
        """ Starting Equilibrium """
        
        title = Tex("{{Comparative Statics}}: how external forces change equilibrium.").set_color_by_tex_to_color_map(
            {"Comparative Statics": YELLOW,}
        ).to_edge(UP)
        self.play(FadeIn(title))
        
        demand_intercept = ValueTracker(100)
        def Demand(q):
            return demand_intercept.get_value() - q/10
        def Inv_Demand(p):
            return (demand_intercept.get_value() - p) * 10
        def Demand_Line():
            return axes.plot(Demand, x_range=[0, Inv_Demand(0)]).set_color(BLUE)        
        demand = always_redraw(Demand_Line)
        
        demand_base = Demand_Line().set_color(GREY)
        demand_base.z_index = -1
        
        supply_intercept = ValueTracker(10)
        def Supply(q):
            return supply_intercept.get_value() + q/10
        def Inv_Supply(p):
            return (supply_intercept.get_value() + p) * 10
        def Supply_Line():
            return axes.plot(Supply, x_range=[0, Inv_Supply(80)]).set_color(YELLOW)        
        supply = always_redraw(Supply_Line)
        
        supply_base = Supply_Line().set_color(GREY)
        supply_base.z_index = -1
        
        def Equilibrium_Price():
            return supply_intercept.get_value() + (demand_intercept.get_value() - supply_intercept.get_value())/2
        
        price_value = ValueTracker(Equilibrium_Price())
        def Price_Line():
            price_value.set_value(Equilibrium_Price())
            return axes.plot(lambda x:price_value.get_value(), x_range=[0,1000]).set_color(RED) 
        price = always_redraw(Price_Line)
        
        price_base = Price_Line().set_color(GREY)
        price_base.z_index = -1
        
        def Quantity_Demanded():
            p = price_value.get_value()
            q = Inv_Demand(p)
            point = axes.c2p(q, p)
            dot = Dot(point).set_color(RED)
            dot.z_index = 3
            line = axes.get_vertical_line(axes.input_to_graph_point(q, demand), color=RED)
            return VGroup(dot, line)
        quantity_demanded = always_redraw(Quantity_Demanded)
        
        def Quantity_Exchanged(p):
            qs = Inv_Supply(p)
            qd = Inv_Demand(p)
            return min(qs, qd)
                
        self.play(FadeIn(demand), FadeIn(demand_base), 
                  FadeIn(supply), FadeIn(supply_base),
                  FadeIn(price), FadeIn(quantity_demanded))
        self.wait()
        
        """ Comparative Statics """
        
        self.play(demand_intercept.animate.set_value(90))
        self.wait()
        self.play(demand_intercept.animate.set_value(70))
        self.wait()
        self.play(demand_intercept.animate.set_value(100))
        self.wait()
        
        self.play(supply_intercept.animate.set_value(20))
        self.wait()
        self.play(supply_intercept.animate.set_value(30))
        self.wait()
        self.play(supply_intercept.animate.set_value(10))
        self.wait()
        
        self.play(supply_intercept.animate.set_value(20))
        self.wait()
        self.play(demand_intercept.animate.set_value(70))
        self.wait()
        
        self.play(supply_intercept.animate.set_value(10), demand_intercept.animate.set_value(100))
        self.wait()
        
        """ Welfare Analysis """
        
        new_title = Tex("{{Welfare Analysis}}: the study of the benefits of markets.").set_color_by_tex_to_color_map(
            {"Welfare Analysis": YELLOW,}
        ).to_edge(UP)
        self.play(Transform(title, new_title))
        
        def Consumer_Surplus():
            area = axes.get_area(Demand_Line(), [0, Quantity_Exchanged(Equilibrium_Price())], bounded_graph=Price_Line(), color=BLUE, opacity=0.5)
            area.z_index = -3
            return area
        CS = always_redraw(Consumer_Surplus)
        CS_base = Consumer_Surplus().set_color(GREY)
        
        self.play(FadeIn(CS), FadeIn(CS_base))
        self.wait()
        
        self.play(demand_intercept.animate.set_value(90))
        self.wait()
        self.play(demand_intercept.animate.set_value(70))
        self.wait()
        self.play(demand_intercept.animate.set_value(100))
        self.wait()
        
        self.play(supply_intercept.animate.set_value(20))
        self.wait()
        self.play(supply_intercept.animate.set_value(30))
        self.wait()
        self.play(supply_intercept.animate.set_value(10))
        self.wait()
        
        self.play(FadeOut(CS), FadeOut(CS_base))
        
        def Producer_Surplus():
            area = axes.get_area(Supply_Line(), [0, Quantity_Exchanged(Equilibrium_Price())], bounded_graph=Price_Line(), color=YELLOW, opacity=0.5)
            area.z_index = -3
            return area
        PS = always_redraw(Producer_Surplus)
        
        PS_base = Producer_Surplus().set_color(GREY)
        
        self.play(FadeIn(PS), FadeIn(PS_base))
        self.wait()
        
        self.play(demand_intercept.animate.set_value(90))
        self.wait()
        self.play(demand_intercept.animate.set_value(70))
        self.wait()
        self.play(demand_intercept.animate.set_value(100))
        self.wait()
        
        self.play(supply_intercept.animate.set_value(20))
        self.wait()
        self.play(supply_intercept.animate.set_value(30))
        self.wait()
        self.play(supply_intercept.animate.set_value(10))
        self.wait()
        
        self.play(FadeOut(PS), FadeOut(PS_base))
        
        def Total_Surplus():
            area = axes.get_area(Supply_Line(), [0, Quantity_Exchanged(Equilibrium_Price())], bounded_graph=Demand_Line(), color=PURPLE, opacity=0.5)
            area.z_index = -3
            return area
        TS = always_redraw(Total_Surplus)
        
        TS_base = Total_Surplus().set_color(GREY)

        self.play(FadeIn(TS), FadeIn(TS_base))
        self.wait()

        self.play(demand_intercept.animate.set_value(90))
        self.wait()
        self.play(demand_intercept.animate.set_value(70))
        self.wait()
        self.play(demand_intercept.animate.set_value(100))
        self.wait()
        
        self.play(supply_intercept.animate.set_value(20))
        self.wait()
        self.play(supply_intercept.animate.set_value(30))
        self.wait()
        self.play(supply_intercept.animate.set_value(10))
        self.wait()
        
        self.play(FadeOut(TS), FadeOut(TS_base))

        """ Price Control """
        
        price_value.suspend_updating()
        
        new_title = Tex(
            "{{Price Controls}}: government price restrictions.",
        ).set_color_by_tex_to_color_map(
            {"Price Controls": YELLOW,}
        ).to_edge(UP)
        self.play(Transform(title, new_title))
        
        self.remove(price)
        
        price_value = ValueTracker(Equilibrium_Price())
        def Price_Line():
            return axes.plot(lambda x:price_value.get_value(), x_range=[0,1000]).set_color(RED) 
        price = always_redraw(Price_Line)
        
        def Quantity_Supplied():
            p = price_value.get_value()
            q = Inv_Supply(p)
            point = axes.c2p(q, p)
            dot = Dot(point).set_color(RED)
            dot.z_index = 3
            line = axes.get_vertical_line(axes.input_to_graph_point(q, supply), color=RED)
            return VGroup(dot, line)
        quantity_supplied = always_redraw(Quantity_Supplied)
        
        self.add(price, quantity_supplied)
        
        self.play(FadeIn(CS), FadeIn(PS), FadeIn(TS_base))
        self.wait()
        
        self.play(price_value.animate.set_value(40))
        
        
```
