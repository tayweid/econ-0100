## Animation 1 | Social Planner

The social planner sets MSC = MSB where the market sets MPC = MPB. When there are no externalities the two solutions are equal and we get the First Welfare Theorem. With externalities they differ. Show Consumer Surplus, then Producer Surplus, then Externalities, then DWL. Show the proceedure to find DWL. Move the value of the externalities around. Show how PS and CS don't move with externalities.


```python
%%manim animation_0

class animation_0(MovingCameraScene):

    def construct(self):
        
        """ Definitions """
        
        PQ_axis = Axes(            
            x_range=[0, 1000, 1000],
            x_length = 7,
            axis_config={"color": WHITE},
            x_axis_config={
                "numbers_to_include": np.arange(0, 1000, 1100),
                "decimal_number_config": {
                    "num_decimal_places":0,
                },
            },
            y_range=[0, 100, 100],
            y_length = 6,
            y_axis_config={
                "numbers_to_include": [100,100,0],
                "decimal_number_config": {
                    "num_decimal_places":0,
                }
            },
            tips=False,
        )
        PQ_axis_small = Axes(            
            x_range=[-1, 1, 1],
            x_length = 1,
            axis_config={"color": WHITE},
            x_axis_config={
                "numbers_to_include": [],
                "decimal_number_config": {
                    "num_decimal_places":0,
                },
            },
            y_range=[-40, 40, 40],
            y_length = 6,
            y_axis_config={
                "numbers_to_include": [],
                "numbers_with_elongated_ticks": [1],
                "decimal_number_config": {
                    "num_decimal_places":2,
                }
            },
            tips=False,
        )
        
        """ Starting Objects """
        
        axes = PQ_axis.copy().scale(0.7).shift(DOWN/3 + RIGHT)

        y_label = axes.get_y_axis_label("P")
        x_label = axes.get_x_axis_label("Q")
        grid_labels = VGroup(x_label, y_label)
        
        part_label = Tex('Part C $|$ Market Failure').set_color(GREY).set_color_by_tex_to_color_map(
            {"Part C": BLUE,}
        ).to_edge(UP, buff=-1/2)
        
        self.play(FadeIn(axes), FadeIn(grid_labels), FadeIn(part_label))
        
        """ Starting Equilibrium """
        
        demand_intercept = ValueTracker(100)
        def Demand(q):
            return demand_intercept.get_value() - q/10
        def Inv_Demand(p):
            return (demand_intercept.get_value() - p) * 10
        def Demand_Line():
            return axes.plot(Demand, x_range=[0, Inv_Demand(0)]).set_color(BLUE)        
        demand = always_redraw(Demand_Line)
        
        demand_base = Demand_Line().set_color(GREY)
        demand_base.z_index = -1
        
        supply_intercept = ValueTracker(0)
        def Supply(q):
            return supply_intercept.get_value() + q/10
        def Inv_Supply(p):
            return (supply_intercept.get_value() + p) * 10
        def Supply_Line():
            return axes.plot(Supply, x_range=[0, Inv_Supply(90)]).set_color(YELLOW)
        supply = always_redraw(Supply_Line)
        
        tax_value = ValueTracker(0)
        
        #def Buyers_Price():
        #def Sellers_Price():
            
        def Equilibrium_Price():
            return supply_intercept.get_value() + (demand_intercept.get_value() - supply_intercept.get_value())/2
        
        def Market():
            p = Equilibrium_Price()
            #p_b = Buyers_Price()
            #p_s = Sellers_Price()
            
            q = Inv_Demand(p)
            #q_d = Inv_Demand(p_b)
            #q_s = Inv_Supply(p_s)
            
            point = axes.c2p(q, p)
            dot = Dot(point).set_color(RED)
            dot.z_index = 3
            qline = axes.get_vertical_line(axes.input_to_graph_point(q, demand), color=RED)
            pline = axes.get_horizontal_line(axes.input_to_graph_point(q, demand), color=RED)
            
            if tax_value == 0:
                Q_hat = Tex('$\hat{Q}$').scale(0.7).next_to(qline, DOWN/2).set_color(RED)
                P_hat = Tex('$\hat{P}$').scale(0.7).next_to(pline, LEFT/2).set_color(RED)
            else:
                Q_hat = Tex('$Q_{T}$').scale(0.7).next_to(qline, DOWN/2).set_color(RED)
                P_hat = VGroup(
                    Tex('$P_b$').scale(0.7).next_to(pline, LEFT/2).set_color(RED),
                    Tex('$P_s$').scale(0.7).next_to(pline, LEFT/2).set_color(RED)
                )
            
            return VGroup(dot, qline, pline, Q_hat, P_hat)
        market = always_redraw(Market)
        
        externality_value = ValueTracker(0)
        externality_value.set_value(0)
        
        def MSC(q):
            adjust = 0
            if externality_value.get_value() < 0:
                adjust = - externality_value.get_value()
            return supply_intercept.get_value() + adjust + q/10
        def Inv_MSC(p):
            adjust = 0
            if externality_value.get_value() < 0:
                adjust = - externality_value.get_value()
            return (supply_intercept.get_value() + adjust + p) * 10
        def MSC_Line():
            line = axes.plot(MSC, x_range=[0, Inv_MSC(90)]).set_color(GREEN)
            line.z_index = -1
            return line
        msc = always_redraw(MSC_Line)
        
        def MSB(q):
            adjust = 0
            if externality_value.get_value() > 0:
                adjust = externality_value.get_value()
            return demand_intercept.get_value() + adjust - q/10
        def Inv_MSB(p):
            adjust = 0
            if externality_value.get_value() > 0:
                adjust = externality_value.get_value()
            return (demand_intercept.get_value() + adjust - p) * 10
        def MSB_Line():
            line = axes.plot(MSB, x_range=[0, Inv_MSB(0)]).set_color(GREEN)
            line.z_index = -1
            return line
        msb = always_redraw(MSB_Line)
        
        
        def Social_Price():
            pos_ext = 0
            neg_ext = 0
            if externality_value.get_value() <= 0:
                neg_ext = - externality_value.get_value()
            if externality_value.get_value() > 0:
                pos_ext = externality_value.get_value()
            return supply_intercept.get_value() + neg_ext + (demand_intercept.get_value() + pos_ext - supply_intercept.get_value() - neg_ext)/2
        
        def Social_Quantity():
            return Inv_MSB(Social_Price())
        
        def Social():
            p = Social_Price()
            q = Social_Quantity()
            point = axes.c2p(q, p)
            dot = Dot(point).set_color(PINK)
            dot.z_index = 3
            qline = axes.get_vertical_line(axes.input_to_graph_point(q, msc), color=PINK)
            #pline = axes.get_horizontal_line(axes.input_to_graph_point(q, msc), color=PINK)
            
            Q_hat = Tex('$Q^*$').scale(0.7).next_to(qline, DOWN*3).set_color(PINK)
            
            return VGroup(dot, qline, Q_hat)
        social = always_redraw(Social)
        
        self.play(FadeIn(demand), FadeIn(supply), FadeIn(msc), FadeIn(msb), FadeIn(market), FadeIn(social))
        self.wait()
        
        def Price():
            return axes.plot(lambda x: Equilibrium_Price(), x_range=[0, Inv_Demand(Equilibrium_Price())])
        price = always_redraw(Price)

        def Quantity_Exchanged(p):
            qs = Inv_Supply(p)
            qd = Inv_Demand(p)
            return min(qs, qd)

        """ Show CS  """
        
        def Consumer_Surplus():
            area = axes.get_area(Demand_Line(), [0, Quantity_Exchanged(Equilibrium_Price())], bounded_graph=price, color=BLUE, opacity=0.5)
            area.z_index = -3
            return area
        CS = always_redraw(Consumer_Surplus)
        
        self.play(FadeIn(CS))
        self.wait()
        
        self.play(externality_value.animate.set_value(-30))
        self.wait()
        
        self.play(externality_value.animate.set_value(0))
        self.wait()
        
        self.play(FadeOut(CS))
        self.wait()
        
        """ Show PS  """
        
        def Producer_Surplus():
            area = axes.get_area(Supply_Line(), [0, Quantity_Exchanged(Equilibrium_Price())], bounded_graph=price, color=YELLOW, opacity=0.5)
            area.z_index = -3
            return area
        PS = always_redraw(Producer_Surplus)
        
        self.play(FadeIn(PS))
        self.wait()
        
        self.play(externality_value.animate.set_value(-30))
        self.wait()
        
        self.play(externality_value.animate.set_value(0))
        self.wait()
        
        self.play(FadeOut(PS))
        self.wait()
        
        """ Show Externality  """
        
        def Externality():
            area = axes.get_area(Supply_Line(), [0, Quantity_Exchanged(Equilibrium_Price())], bounded_graph=msc, color=GREEN, opacity=0.5)
            area.z_index = -3
            return area
        EXT = always_redraw(Externality)
        
        label = Tex('EXT').set_color(GREEN).to_edge(UP+LEFT, buff=0.1).shift(DOWN*2)
        self.play(FadeIn(label))
        
        definitions = [
            Tex('- Between MB and MSB').set_color(GREEN).scale(0.7).to_edge(UP+LEFT, buff=0.1).shift(DOWN*3),
            Tex('- Between MC and MSC').set_color(GREEN).scale(0.7).to_edge(UP+LEFT, buff=0.1).shift(DOWN*3.5),
            Tex('- Inside $Q^*$').set_color(GREEN).scale(0.7).to_edge(UP+LEFT, buff=0.1).shift(DOWN*4),
        ]
        for d in definitions:
            self.play(FadeIn(d))
            self.wait()
        
        self.play(FadeIn(EXT))
        self.wait()
        
        self.play(externality_value.animate.set_value(-30))
        self.wait()
        
        self.play(externality_value.animate.set_value(0))
        self.wait()
        
        self.play(FadeOut(EXT))
        self.wait()
        
        """ Show DWL  """

        def DeadWeightLoss():
            area = axes.get_area(MSC_Line(), [Social_Quantity(), Quantity_Exchanged(Equilibrium_Price())], bounded_graph=msb, color=GREY, opacity=0.5)
            area.z_index = -3
            return area
        DWL = always_redraw(DeadWeightLoss)
        
        new_label = Tex('DWL').set_color(GREY).to_edge(UP+LEFT, buff=0.1).shift(DOWN*2)
        self.play(Transform(label, new_label), *[FadeOut(d) for d in definitions])
        
        definitions = [
            Tex('- Between MSC and MSB').set_color(GREY).scale(0.7).to_edge(UP+LEFT, buff=0.1).shift(DOWN*3),
            Tex('- Between $\hat{Q}$ and $Q^*$').set_color(GREY).scale(0.7).to_edge(UP+LEFT, buff=0.1).shift(DOWN*3.5),
        ]
        for d in definitions:
            self.play(FadeIn(d))
            self.wait()
        
        self.play(externality_value.animate.set_value(-30))
        self.wait()
        
        self.play(FadeIn(DWL))
        self.wait()
        
        self.play(externality_value.animate.set_value(0))
        self.wait()
        
        self.play(externality_value.animate.set_value(30))
        self.wait()
        
        self.play(externality_value.animate.set_value(0))
        self.wait()
        
        """ Corrective Taxes """
        
        axes_small = PQ_axis_small.copy().scale(0.7).to_edge(LEFT).shift(DOWN/3 + RIGHT)
        self.play(FadeOut(label), *[FadeOut(d) for d in definitions], FadeIn(axes_small), axes.animate.shift(RIGHT))
        self.wait()
        
        price = ValueTracker(1/2)
        def PriceLine(x):
            return price.get_value()
        def PricePlot():
            line = axes_small.plot(PriceLine, x_range=[-1,0.5]).set_color(RED)
            number = DecimalNumber(num_decimal_places=2).set_color(RED).scale(0.8).next_to(line,LEFT).set_value(price.get_value())
            label = Tex("Price").set_color(RED).next_to(number, LEFT, buff=1/2)

            return VGroup(line, label, number)
        price_line = always_redraw(PricePlot)
        self.play(FadeIn(price_line))
        self.wait()
        
```

```python
%%manim ExternalityTax #-low_quality

class ExternalityTax(GraphScene):
    CONFIG = {
        "x_axis_label": "$Q$",
        "y_axis_label": "$P$",
        "x_min": 0,
        "x_max": 100,
        "y_min": 0,
        "y_max": 10,
        "x_axis_width": FRAME_HEIGHT,
        "y_axis_height":FRAME_HEIGHT / 2,
        "x_labeled_nums": [],
        "y_labeled_nums": [2,10],
        "y_tick_frequency": 10,
        "x_tick_frequency": 100,
        "graph_origin": np.array((-FRAME_X_RADIUS + 1.5*LARGE_BUFF, -FRAME_Y_RADIUS + 1.4*LARGE_BUFF, 0)),
        "demand_color": BLUE,
        "supply_color": ORANGE,
        "expenditure_color": GREEN,
        "cs_color": PINK,
        "ps_color": YELLOW,
        "surplus_color": BLUE,
        "pq_color": RED,
        "axes_color": GREY,
        "dwl_color": GREY,
        "gov_color": RED,
        "ext_color": GREEN,
    }

    def construct(self):
        self.intro_sequence()
        self.wait(3)
        self.initialize_eq()
        self.wait(3)
        self.externality(2,0)
        self.wait(10)
        self.prices_in(2,0)
        self.wait(3)
        self.tax_wedge(2,2)
        self.wait(3)
        self.welfare_in(2,2)
        self.wait(3)
        self.welfare_up(2,0,0)
        self.wait(3)
        self.welfare_up(2,0,4)
        self.wait(3)
        self.welfare_up(4,0,4)
        self.wait(3)
        self.welfare_up(0,2,0)
        self.wait(3)
        self.welfare_up(0,2,-2)
        self.wait(3)
        self.welfare_up(2,0,-2)
        self.wait(3)
        self.welfare_up(2,0,0)
        self.wait(3)
        self.welfare_up(4,0,2)
        self.wait(3)
        self.welfare_up(4,0,4)
        self.wait(3)
        self.welfare_up(2,0,2)

            
    def MPBq(self, priceD):
        return 100 - 10*priceD
    
    def MPBp(self,quantityD):
        return (10 - quantityD/10)
    
    def MPCq(self, priceS):
        return 10*priceS - 20
    
    def MPCp(self,quantityS):
        return 2 + quantityS /10
                
    def intro_sequence(self):
        title = TextMobject("Tutorial 4.2 | Corrective Policies").scale(1.5)
        self.play(FadeIn(title))
        self.wait(1)
        transform_title = TextMobject("Tutorial 4.2 | Corrective Policies").scale(1.2)
        transform_title.to_corner(UP)
        self.play(
            Transform(title, transform_title))
        self.wait(3)
        self.setup_axes(animate=True)
        
    def initialize_eq(self):
        # Introduce the mobjects for the entire scene and graph the private equilibrium

        self.MPB = self.get_graph(self.MPBp, self.demand_color, x_min=0, x_max=100)
        self.MPBeq = self.get_graph(self.MPBp, self.demand_color, x_min=0, x_max=100)
        self.MPBeql = TextMobject("MPB").scale(0.8)
        self.MPBeql.set_color(self.demand_color)
        self.MPBeql.move_to(self.coords_to_point(83, 1),LEFT)
        self.MSB = self.get_graph(self.MPBp, self.demand_color, x_min=0, x_max=100)

        self.MPC = self.get_graph(self.MPCp, self.supply_color, x_min=0, x_max=80)
        self.MPCeq = self.get_graph(self.MPCp, self.supply_color, x_min=0, x_max=80)
        self.MPCeql = TextMobject("MPC").scale(0.8)
        self.MPCeql.set_color(self.supply_color)
        self.MPCeql.move_to(self.coords_to_point(83, 10),LEFT)
        self.MSC = self.get_graph(self.MPCp, self.supply_color, x_min=0, x_max=80)
        
        self.socL = TextMobject("MS").scale(0.8)
        self.socL.set_color(self.demand_color)
        self.socL.move_to(self.coords_to_point(83, 1),LEFT)
        
        eqp1 = self.coords_to_point(0, 6)
        eqp2 = self.coords_to_point(40, 6)
        eqp3 = self.coords_to_point(40, 0)
        
        self.horzLineEq = DashedVMobject(Line(eqp1,eqp2, color=self.axes_color))
        self.priceEq = TextMobject("$P_M^*=$ "+str(6)).scale(0.7)
        self.priceEq.next_to(self.horzLineEq, LEFT)
        self.vertLineEq = DashedVMobject(Line(eqp3,eqp2, color=self.axes_color))
        self.quantEq = TextMobject(str(40)).scale(0.7)
        self.quantEq.next_to(self.vertLineEq, DOWN)

        self.vertLine = DashedVMobject(Line(eqp3,eqp2, color=self.axes_color))
        self.quantDesc = TextMobject(str(40)).scale(0.7)
        self.quantDesc.add_updater(lambda d: d.next_to(self.vertLine, DOWN))
        
        self.vertLineT = DashedVMobject(Line(eqp3,eqp2, color=self.axes_color))
        self.quantDescT = TextMobject(str(40)).scale(0.7)
        self.quantDescT.next_to(self.vertLineT, DOWN)
        
        self.horzLineA =  DashedVMobject(Line(eqp1,eqp2, color=self.axes_color))
        self.priceAL = TextMobject(str(6)).scale(0.7)
        self.priceAL.next_to(self.horzLineA, LEFT)
        
        self.horzLineS =  DashedVMobject(Line(eqp1,eqp2, color=self.axes_color))
        self.priceSL = TextMobject(str(6)).scale(0.7)
        self.priceSL.next_to(self.horzLineS, LEFT)
        
        self.horzLineB =  DashedVMobject(Line(eqp1,eqp2, color=self.axes_color))
        self.priceBL = TextMobject(str(6)).scale(0.7)
        self.priceBL.next_to(self.horzLineB, LEFT)
        
        self.horzLineT =  DashedVMobject(Line(eqp1,eqp2, color=self.pq_color))
        self.priceTL = TextMobject(str(6)).scale(0.7)
        self.priceTL.next_to(self.horzLineT, LEFT)
        
        wp = self.coords_to_point(-6, 6)
        self.taxWedgeS = Line(wp,wp, color=self.axes_color)
        self.taxWedgeSL = TextMobject("Tax").scale(0.7)
        self.taxWedgeSL.next_to(self.taxWedgeS,LEFT)
        
        ep = self.coords_to_point(88, 5)
        self.extWedge = Line(ep,ep, color=self.axes_color)
        self.extWedgeL = TextMobject("Exernal").scale(0.7)
        self.extWedgeL.next_to(self.extWedge,RIGHT)
        
        qL = self.coords_to_point(0, -1.8)
        qU = self.coords_to_point(6, -1.8)
        
        self.quantityExchanged = Line(qL,qU, color=self.demand_color)
        self.quantityExchLabel = TextMobject("Quant. Ex.").scale(0.6)
        self.quantityExchLabel.set_color(self.demand_color)
        self.quantityExchLabel.add_updater(lambda d: d.next_to(self.quantityExchanged,UP))
        
        qL = self.coords_to_point(0, -1.5)
        qU = self.coords_to_point(6, -1.5)
        
        self.subQuant = Line(qL,qU, color=self.axes_color)
        self.subQuantL = TextMobject("Sub. Q.").scale(0.5)
        self.subQuantL.add_updater(lambda d: d.next_to(self.subQuant,DOWN))
        
        self.priceUS = Dot(self.coords_to_point(40, 6))
        self.priceLS = Dot(self.coords_to_point(40, 6))
        
        # WELFARE
        
        a = self.coords_to_point(135, 10)
        b = self.coords_to_point(135, 9)
        self.csL = Line(a,b, color=self.cs_color)
        
        self.csD = TextMobject("CS")
        self.csD.next_to(self.csL,LEFT)
        self.csD.set_color(self.cs_color)
        
        a = self.coords_to_point(135, 8.5)
        b = self.coords_to_point(135, 7.5)
        self.psL = Line(a,b, color=self.ps_color)
        
        self.psD = TextMobject("PS")
        self.psD.next_to(self.psL,LEFT)
        self.psD.set_color(self.ps_color)
        
        a = self.coords_to_point(135, 7)
        b = self.coords_to_point(135, 6)
        self.extL = Line(a,b, color=self.ext_color)
        
        self.extD = TextMobject("EXT")
        self.extD.next_to(self.extL,LEFT)
        self.extD.set_color(self.ext_color)
        
        a = self.coords_to_point(135, 5.5)
        b = self.coords_to_point(135, 4.5)
        self.dwlL = Line(a,b, color=self.dwl_color)
        
        self.dwlD = TextMobject("DWL")
        self.dwlD.next_to(self.dwlL,LEFT)
        self.dwlD.set_color(self.dwl_color)
        
        a = self.coords_to_point(135, 4)
        b = self.coords_to_point(135, 3)
        self.govL = Line(a,b, color=self.gov_color)
        
        self.govD = TextMobject("GOV")
        self.govD.next_to(self.govL,LEFT)
        self.govD.set_color(self.gov_color)

        
        self.play(ShowCreation(self.MPBeq), ShowCreation(self.MPBeql), ShowCreation(self.MSB))
        self.play(ShowCreation(self.MPCeq), ShowCreation(self.MPCeql), ShowCreation(self.MSC))
        self.play(ShowCreation(self.horzLineEq),ShowCreation(self.priceEq))
        self.play(ShowCreation(self.vertLineEq),ShowCreation(self.quantEq))
        
        
    def externality(self, exC, exB):
        # Introduce the externality and the new intersection points
        
        def MSBq(priceD):
            return (10 - exB)*10 - 10*priceD
        def MSBp(quantD):
            return 10 - quantD/10 + exB
        def MSCq(priceS):
            return 10*priceS - 10*(2+exC)
        def MSCp(quantS):
            return 2 + quantS /10 + exC
        
        MSB = DashedVMobject(self.get_graph(MSBp, self.demand_color, x_min=0, x_max=100))
        MSBl = TextMobject("MSB").scale(0.8)
        MSBl.set_color(self.demand_color)
        MSBl.add_updater(lambda d: d.move_to(self.coords_to_point(83, 3+exB),LEFT))
        MSC = DashedVMobject(self.get_graph(MSCp, self.supply_color, x_min=0, x_max=80))
        MSCl = TextMobject("MSC").scale(0.8)
        MSCl.set_color(self.supply_color)
        MSCl.add_updater(lambda d: d.move_to(self.coords_to_point(83, 10+exC),LEFT))
        
        eqQ = 40-5*(exC+exB)
        priceA = MSCp(eqQ)
        priceB = MSCp(40)
        eqPS = self.MPCp(eqQ)
        eqPB = self.MPBp(eqQ)
        
        a = self.coords_to_point(40, 0)
        b = self.coords_to_point(40, priceB)
        vertLineEq = DashedVMobject(Line(a,b, color=self.axes_color))
        quantDescEq = TextMobject(str(40)).scale(0.7)
        quantDescEq.add_updater(lambda d: d.next_to(vertLineEq, DOWN))
        
        a = self.coords_to_point(eqQ, 0)
        b = self.coords_to_point(eqQ, eqPB)
        vertLine = DashedVMobject(Line(a,b, color=self.pq_color))
        quantDesc = TextMobject(str(eqQ)).scale(0.7)
        quantDesc.add_updater(lambda d: d.next_to(vertLine, DOWN))
        
        a = self.coords_to_point(0, priceA)
        b = self.coords_to_point(eqQ, priceA)
        horzLineA = DashedVMobject(Line(a,b, color=self.pq_color))
        priceAL = TextMobject(str(int(priceA))).scale(0.7)
        priceAL.add_updater(lambda d: d.next_to(horzLineA, LEFT))

        a = self.coords_to_point(0, priceB)
        b = self.coords_to_point(40, priceB)
        horzLineB = DashedVMobject(Line(a,b, color=self.axes_color))
        priceBL = TextMobject(str(int(priceB))).scale(0.7)
        priceBL.add_updater(lambda d: d.next_to(horzLineB, LEFT))
        
        qL = self.coords_to_point(40, -2)
        qU = self.coords_to_point(eqQ, -2)
        subQuant = Line(qL,qU, color=self.axes_color)
        
        ewp1 = self.coords_to_point(96, 1)
        ewp2 = self.coords_to_point(96, 1+exB)
        if exC > 0:
            ewp1 = self.coords_to_point(96, 10)
            ewp2 = self.coords_to_point(96, 10+exC)
        extWedge = Line(ewp1,ewp2, color=self.pq_color)
        extWedgeL = TextMobject("Ex.="+str(abs(exC)+abs(exB))).scale(0.7)
        extWedgeL.next_to(extWedge, RIGHT)

        self.play(Transform(self.MSC,MSC), Transform(self.MSB,MSB),
                  Transform(self.extWedge,extWedge), Transform(self.extWedgeL,extWedgeL))
        
        self.wait(2)
        
        if exC > 0:
            self.play(Transform(self.socL,MSCl))
        if exB > 0:
            self.play(Transform(self.socL,MSBl))
            
        self.wait(2)
        
        self.play(Transform(self.quantDesc,quantDesc), Transform(self.vertLine,vertLine),
                  
                  #Transform(self.vertLineEq,vertLineEq), 
                  Transform(self.horzLineA,horzLineA),
                  #Transform(self.horzLineB,horzLineB),
                  Transform(self.priceAL,priceAL), #Transform(self.priceB,priceB),
                  Transform(self.subQuant,subQuant))
        self.add(self.subQuantL)
        
    def prices_in(self, exC, exB):
        # Introduce excess and shortage at a few prices
        def MSBq(priceD):
            return (10 - exB)*10 - 10*priceD
        def MSBp(quantD):
            return 10 - quantD/10 + exB
        def MSCq(priceS):
            return 10*priceS - 10*(2+exC)
        def MSCp(quantS):
            return 2 + quantS /10 + exC
        
        eqQ = 40-5*(exC+exB)
        eqPS = self.MPCp(eqQ)
        eqPB = self.MPBp(eqQ)
        
        priceU = max(eqPS, eqPB)
        priceL = min(eqPS, eqPB)
        
        priceUS = Dot(self.coords_to_point(eqQ, priceU))
        priceLS = Dot(self.coords_to_point(eqQ, priceL))
        
        for price in [priceU, priceL]:
            quant = max(self.MPCq(price), self.MPBq(price))
        
            a = self.coords_to_point(0, price)
            b = self.coords_to_point(quant, price)
            horzLineT = Line(a,b, color=self.pq_color)
            priceTL = TextMobject(str(int(price))).scale(0.7)
            priceTL.next_to(horzLineT, LEFT)

            a = self.coords_to_point(quant, 0)
            b = self.coords_to_point(quant, price)
            vertLineT = DashedVMobject(Line(a,b, color=self.pq_color))
            quantDescT = TextMobject(str(int(quant))).scale(0.7)
            quantDescT.next_to(vertLineT, DOWN)
            
            qL = self.coords_to_point(eqQ, -2)
            qU = self.coords_to_point(quant, -2)
            subQuant = Line(qL,qU, color=self.demand_color) # This is shortage, excess this scene
            if price < 6:
                subQuantL = TextMobject("Shortage").scale(0.5)
            if price > 6:
                subQuantL = TextMobject("Excess").scale(0.5)
            subQuantL.next_to(subQuant,DOWN)
            
            self.remove(self.subQuantL)
            
            self.add(priceUS)
            self.add(priceLS)

            self.play(Transform(self.horzLineT,horzLineT), Transform(self.priceTL,priceTL),
                      Transform(self.vertLineT,vertLineT), Transform(self.quantDescT,quantDescT),
                      Transform(self.subQuant,subQuant), Transform(self.subQuantL,subQuantL))
            self.wait(10)
        
        qL = self.coords_to_point(40, -2)
        qU = self.coords_to_point(eqQ, -2)
        subQuant = Line(qL,qU, color=self.axes_color)
        subQuantL = TextMobject("Sub. Q.").scale(0.5)
        subQuantL.next_to(subQuant,DOWN)
        
        self.play(FadeOut(self.horzLineT), FadeOut(self.priceTL),
                  FadeOut(self.vertLineT), FadeOut(self.quantDescT),
                  Transform(self.subQuant,subQuant), Transform(self.subQuantL,subQuantL))
        self.remove(priceUS)
        self.remove(priceLS)
        
        
    def tax_wedge(self,tax,ext):
        self.play(FadeOut(self.priceEq))
        
        quant = int(40-5*(tax))
        quantS = 40-5*(ext)
        priceS = int(self.MPCp(quant))
        priceB = int(self.MPBp(quant))
        
        twp1 = self.coords_to_point(-6, priceS)
        twp2 = self.coords_to_point(-6, priceB)

        taxWedgeS = Line(twp1,twp2, color=self.pq_color)
        taxWedgeSL = TextMobject("Tax").scale(0.7)
        taxWedgeSL.next_to(taxWedgeS, LEFT)
        
        eqpt1 = self.coords_to_point(quant, 0)
        eqpt2 = self.coords_to_point(quant, priceS)
        eqpt3 = self.coords_to_point(0, priceS)
        eqpt4 = self.coords_to_point(quant, priceB)
        eqpt5 = self.coords_to_point(0, priceB)

        vertLine = DashedVMobject(Line(eqpt1,eqpt4, color=self.pq_color))
        quantDesc = TextMobject(str(quant)).scale(0.7)
        quantDesc.next_to(vertLine, DOWN)
        
        horzLineS = DashedVMobject(Line(eqpt2,eqpt3, color=self.pq_color))
        priceSL = TextMobject(str(priceS)).scale(0.7)
        priceSL.next_to(horzLineS, LEFT)
        
        horzLineB = DashedVMobject(Line(eqpt4,eqpt5, color=self.pq_color))
        priceBL = TextMobject(str(priceB)).scale(0.7)
        priceBL.next_to(horzLineB, LEFT)
        
        qL = self.coords_to_point(0, -1.8)
        qU = self.coords_to_point(quant, -1.8)
        quantityExchanged = Line(qL,qU, color=self.demand_color)
        
        qL = self.coords_to_point(quant, -2)
        qU = self.coords_to_point(quantS, -2)
        subQuant = Line(qL,qU, color=self.axes_color)
        
        self.add(self.quantityExchLabel)
        
        self.play(Transform(self.taxWedgeS,taxWedgeS), Transform(self.taxWedgeSL,taxWedgeSL),
                    Transform(self.quantDesc,quantDesc), Transform(self.subQuant,subQuant),
                    Transform(self.vertLine,vertLine), Transform(self.quantityExchanged,quantityExchanged),
                    Transform(self.horzLineS,horzLineS), Transform(self.priceSL,priceSL),
                    Transform(self.horzLineB,horzLineB), Transform(self.priceBL,priceBL))
        
        
        
    def welfare_in(self, ext, tax):
        # Introduce welfare with externalities
        
        def MSCq(priceS):
            return 10*priceS - 10*(2+ext)
        def MSCp(quantS):
            return 2 + quantS /10 + ext
        
        quant = 40-5*(tax)
        quantS = 40-5*(ext)
        
        priceB = self.MPBp(quant)
        priceS = self.MPCp(quant)
                
        priceU = MSCp(quantS)
        priceL = self.MPBp(quantS)
        
        qL = self.coords_to_point(quant, -2)
        qU = self.coords_to_point(quantS, -2)
        subQuant = Line(qL,qU, color=self.axes_color)
        self.play(Transform(self.subQuant,subQuant))
        
        # CS
        a = self.coords_to_point(0, 10)
        b = self.coords_to_point(quant, priceB)
        c = self.coords_to_point(0, priceB)
        self.cs = Polygon(a,b,c, fill_opacity = 0.2, fill_color = self.cs_color, color=self.cs_color)
        self.csN = Polygon(a,b,c, fill_opacity = 0.2, fill_color = self.cs_color, color=self.cs_color)
        
        csN = TextMobject(str(int((10-priceB)*quant/2)))
        csN.next_to(self.csL, RIGHT)
        csN.set_color(self.cs_color)
        
        self.add(self.csL,self.csD)
        self.play(FadeIn(self.cs),Transform(self.csN,csN))
        self.wait(6)
        
        # PS
        a = self.coords_to_point(0, priceS)
        b = self.coords_to_point(quant, priceS)
        c = self.coords_to_point(0, 2)
        self.ps = Polygon(a,b,c, fill_opacity = 0.2, fill_color = self.ps_color, color=self.ps_color)
        self.psN = Polygon(a,b,c, fill_opacity = 0.2, fill_color = self.ps_color, color=self.ps_color)
        
        psN = TextMobject(str(int((priceS-2)*quant/2)))
        psN.next_to(self.psL, RIGHT)
        psN.set_color(self.ps_color)
        
        self.add(self.psL,self.psD)
        self.play(FadeIn(self.ps),Transform(self.psN,psN))
        self.wait(6)

        # EXT
        a = self.coords_to_point(0,2)
        b = self.coords_to_point(0,2+ext)
        c = self.coords_to_point(quant,priceB)
        d = self.coords_to_point(quant,priceS)
        self.ext = Polygon(a,b,c,d, fill_opacity = 0.2, fill_color = self.ext_color, color=self.ext_color)
        self.extN = Polygon(a,b,c,d, fill_opacity = 0.2, fill_color = self.ext_color, color=self.ext_color)

        extN = TextMobject(str(int(ext*quant)))
        extN.next_to(self.extL, RIGHT)
        extN.set_color(self.ext_color)

        self.add(self.extL,self.extD)
        self.play(FadeIn(self.ext),Transform(self.extN,extN))
        self.wait(6)
        
        # DWL
        a = self.coords_to_point(quant,priceU)
        b = self.coords_to_point(quantS,priceB)
        c = self.coords_to_point(quant,priceL)
        self.dwl = Polygon(a,b,c, fill_opacity = 1, fill_color = self.dwl_color, color=self.dwl_color)
        self.dwlN = Polygon(a,b,c, fill_opacity = 1, fill_color = self.dwl_color, color=self.dwl_color)
        
        dwlN = TextMobject(str(int((priceB-priceS)*abs(quantS-quant)/2)))
        dwlN.next_to(self.dwlL, RIGHT)
        dwlN.set_color(self.dwl_color)
        
        self.add(self.dwlL,self.dwlD)
        self.play(FadeIn(self.dwl),Transform(self.dwlN,dwlN))
        self.wait(6)
        
        # GOV
        a = self.coords_to_point(0,priceB)
        b = self.coords_to_point(0,priceS)
        c = self.coords_to_point(quant,priceS)
        d = self.coords_to_point(quant,priceB)
        self.gov = Polygon(a,b,c,d, fill_opacity = 0.2, fill_color = self.gov_color, color=self.gov_color)
        self.govN = Polygon(a,b,c,d, fill_opacity = 0.2, fill_color = self.gov_color, color=self.gov_color)

        govN = TextMobject(str(int(tax*quant)))
        govN.next_to(self.govL, RIGHT)
        govN.set_color(self.gov_color)

        self.add(self.govL,self.govD)
        self.play(FadeIn(self.gov),Transform(self.govN,govN))  
        
    def welfare_up(self, exC, exB, tax):
        # Update externality, taxes, and welfare
        ex = exC-exB
        def MSBq(priceD):
            return (10 - exB)*10 - 10*priceD
        def MSBp(quantD):
            return 10 - quantD/10 + exB
        def MSCq(priceS):
            return 10*priceS - 10*(2+exC)
        def MSCp(quantS):
            return 2 + quantS /10 + exC
        
        MSB = DashedVMobject(self.get_graph(MSBp, self.demand_color, x_min=0, x_max=100))
        MSC = DashedVMobject(self.get_graph(MSCp, self.supply_color, x_min=0, x_max=80))
        
        if exC > 0:
            socL = TextMobject("MSC").scale(0.8)
            socL.move_to(self.coords_to_point(83, 10+exC),LEFT)
            socL.set_color(self.supply_color)
        if exB > 0:
            socL = TextMobject("MSB").scale(0.8)
            socL.move_to(self.coords_to_point(83, 1+exB),LEFT)
            socL.set_color(self.demand_color)
        if exC+exB == 0:
            socL = TextMobject("MPC").scale(0.8)
            socL.move_to(self.coords_to_point(83, 10),LEFT)
            socL.set_color(self.supply_color)
        
        quant = 40-5*(tax)
        quantS = 40-5*(ex)

        priceB = self.MPBp(quant)
        priceS = self.MPCp(quant)
        
        priceU = max(priceB,priceS,MSCp(quant),MSBp(quant))
        priceL = min(priceB,priceS,MSCp(quant),MSBp(quant))
        
        twp1 = self.coords_to_point(-6, priceS)
        twp2 = self.coords_to_point(-6, priceB)

        taxWedgeS = Line(twp1,twp2, color=self.pq_color)
        taxWedgeSL = TextMobject("Tax").scale(0.7)
        taxWedgeSL.next_to(taxWedgeS, LEFT)
        
        eqpt1 = self.coords_to_point(quant, 0)
        eqpt2 = self.coords_to_point(quant, priceS)
        eqpt3 = self.coords_to_point(0, priceS)
        eqpt4 = self.coords_to_point(quant, priceB)
        eqpt5 = self.coords_to_point(0, priceB)
        
        a = self.coords_to_point(quant, 0)
        b = self.coords_to_point(quant, priceU)
        vertLine = DashedVMobject(Line(a,b, color=self.pq_color))
        quantDesc = TextMobject(str(int(quant))).scale(0.7)
        quantDesc.next_to(vertLine, DOWN)
        
        a = self.coords_to_point(0, priceU)
        b = self.coords_to_point(quant, priceU)
        horzLineA = DashedVMobject(Line(a,b, color=self.pq_color))
        priceAL = TextMobject(str(int(priceU))).scale(0.7)
        priceAL.next_to(horzLineA, LEFT)
        
        horzLineS = DashedVMobject(Line(eqpt2,eqpt3, color=self.pq_color))
        priceSL = TextMobject(str(int(priceS))).scale(0.7)
        priceSL.next_to(horzLineS, LEFT)
        
        horzLineB = DashedVMobject(Line(eqpt4,eqpt5, color=self.pq_color))
        priceBL = TextMobject(str(int(priceB))).scale(0.7)
        priceBL.next_to(horzLineB, LEFT)
        
        qL = self.coords_to_point(quant, -2)
        qU = self.coords_to_point(quantS, -2)
        subQuant = Line(qL,qU, color=self.axes_color)
        
        qL = self.coords_to_point(0, -1.8)
        qU = self.coords_to_point(quant, -1.8)
        quantityExchanged = Line(qL,qU, color=self.demand_color)
        
        # CS
        a = self.coords_to_point(0, 10)
        b = self.coords_to_point(quant, priceB)
        c = self.coords_to_point(0, priceB)
        cs = Polygon(a,b,c, fill_opacity = 0.2, fill_color = self.cs_color, color=self.cs_color)
        csN = TextMobject(str(int((10-priceB)*quant/2)))
        csN.next_to(self.csL, RIGHT)
        csN.set_color(self.cs_color)
        
        # PS
        a = self.coords_to_point(0, priceS)
        b = self.coords_to_point(quant, priceS)
        c = self.coords_to_point(0, 2)
        ps = Polygon(a,b,c, fill_opacity = 0.2, fill_color = self.ps_color, color=self.ps_color)
        psN = TextMobject(str(int((priceS-2)*quant/2)))
        psN.next_to(self.psL, RIGHT)
        psN.set_color(self.ps_color)

        # EXT
        a = self.coords_to_point(0,2)
        b = self.coords_to_point(0,2+ex)
        c = self.coords_to_point(quant,MSCp(quant))
        d = self.coords_to_point(quant,self.MPCp(quant))
        if exB > 0:
            a = self.coords_to_point(0,10+exB)
            b = self.coords_to_point(0,10)
            c = self.coords_to_point(quant,self.MPBp(quant))
            d = self.coords_to_point(quant,MSBp(quant))
        ext = Polygon(a,b,c,d, fill_opacity = 0.2, fill_color = self.ext_color, color=self.ext_color)
        extN = TextMobject(str(abs(int(ex*quant))))
        extN.next_to(self.extL, RIGHT)
        extN.set_color(self.ext_color)
        
        # DWL
        if quantS >= quant:
            a = self.coords_to_point(quant,MSBp(quant))
            b = self.coords_to_point(quantS,MSBp(quantS))
            c = self.coords_to_point(quant,MSCp(quant))
        if quantS < quant:
            a = self.coords_to_point(quant,MSCp(quant))
            b = self.coords_to_point(quantS,MSBp(quantS))
            c = self.coords_to_point(quant,MSBp(quant))
        dwl = Polygon(a,b,c, fill_opacity = 1, fill_color = self.dwl_color, color=self.dwl_color)
        dwlN = TextMobject(str(int((priceU-priceL)*abs(quantS-quant)/2)))
        dwlN.next_to(self.dwlL, RIGHT)
        dwlN.set_color(self.dwl_color)
        
        # GOV
        a = self.coords_to_point(0,priceB)
        b = self.coords_to_point(0,priceS)
        c = self.coords_to_point(quant,priceS)
        d = self.coords_to_point(quant,priceB)
        gov = Polygon(a,b,c,d, fill_opacity = 0.2, fill_color = self.gov_color, color=self.gov_color)
        govN = TextMobject(str(int(tax*quant)))
        govN.next_to(self.govL, RIGHT)
        govN.set_color(self.gov_color)
        
        ewp1 = self.coords_to_point(96, 1)
        ewp2 = self.coords_to_point(96, 1+exB)
        if exC > 0:
            ewp1 = self.coords_to_point(96, 10)
            ewp2 = self.coords_to_point(96, 10+exC)
        extWedge = Line(ewp1,ewp2, color=self.pq_color)
        extWedgeL = TextMobject("Ex.="+str(abs(exC)+abs(exB))).scale(0.7)
        extWedgeL.next_to(extWedge, RIGHT)
        
        self.play(Transform(self.taxWedgeS,taxWedgeS), Transform(self.taxWedgeSL,taxWedgeSL),
                    Transform(self.quantDesc,quantDesc), Transform(self.subQuant,subQuant),
                    Transform(self.vertLine,vertLine), Transform(self.quantityExchanged,quantityExchanged),
                    Transform(self.horzLineS,horzLineS), Transform(self.priceSL,priceSL),
                    Transform(self.horzLineB,horzLineB), Transform(self.priceBL,priceBL),
                    Transform(self.horzLineA,horzLineA), Transform(self.priceAL,priceAL),
                  
                    Transform(self.socL,socL), Transform(self.subQuant,subQuant),
                  
                    Transform(self.MSC,MSC), Transform(self.MSB,MSB),
                  
                    Transform(self.gov,gov),Transform(self.govN,govN),
                    Transform(self.cs,cs),Transform(self.csN,csN),
                    Transform(self.ps,ps),Transform(self.psN,psN),
                    Transform(self.ext,ext),Transform(self.extN,extN),
                    Transform(self.dwl,dwl),Transform(self.dwlN,dwlN),
                    Transform(self.extWedge,extWedge), Transform(self.extWedgeL,extWedgeL))
        
    def summary(self):
        self.play(FadeOut(self.psN), FadeOut(self.psL), FadeOut(self.psD),
                  FadeOut(self.csN), FadeOut(self.csL), FadeOut(self.csD),
                  FadeOut(self.dwlN), FadeOut(self.dwlL), FadeOut(self.dwlD),
                  FadeOut(self.extN), FadeOut(self.extL), FadeOut(self.extD))
        
        reading = TextMobject("Mankiw Ch. 12 for more info")
        reading.move_to(self.coords_to_point(60, 12), LEFT)
        self.play(FadeIn(reading))
```
