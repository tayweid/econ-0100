```python
import matplotlib.pyplot as plt
import numpy as np

# Set up the figure and axis with specific dimensions
plt.figure(figsize=(6, 6), dpi=100)

# Define the origin coordinates
x_origin = 1/4  # Adjust this value to set the x-coordinate of the origin
y_origin = 1/4  # Adjust this value to set the y-coordinate of the origin

vertical_offset = 1


# Set limits for x and y axes
plt.xlim(0, 10 + vertical_offset)
plt.ylim(0, 10 + vertical_offset)

# Draw the x-axis and y-axis
plt.plot([0, 10 + vertical_offset], [y_origin, y_origin], color='black', linewidth=1)  # X-axis line
plt.plot([x_origin, x_origin], [0, 10 + vertical_offset], color='black', linewidth=1)  # Y-axis line

# Remove tick marks and labels
plt.xticks([])
plt.yticks([])

# Define supply and demand curves
x = np.linspace(x_origin, 10, 100)
demand_intercept = 10 + vertical_offset
demand = demand_intercept - x  # Demand curve: downward sloping
supply = vertical_offset + x   # Supply curve: upward sloping

# Plot supply and demand curves
plt.plot(x, demand, color='black', label='Demand')
plt.plot(x, supply, color='black', label='Supply')

# Equilibrium price and quantity
equilibrium_quantity = 5
equilibrium_price = 5 + vertical_offset

# Plot dashed lines for equilibrium price and quantity
plt.axhline(y=equilibrium_price, color='black', linestyle=':', xmin=0.02, xmax=0.45)
plt.axvline(x=equilibrium_quantity, color='black', linestyle=':', ymin=0.02, ymax=0.55)  # Vertical line at equilibrium quantity

# Add labels for consumer surplus (A) and producer surplus (B)
plt.text(equilibrium_quantity / 4, equilibrium_price + (demand_intercept - equilibrium_price) / 4, 
         'A', fontsize=12, color='black')  # Label for consumer surplus (A)
plt.text(equilibrium_quantity / 4, equilibrium_price - (demand_intercept - equilibrium_price) / 4, 
         'B', fontsize=12, color='black')    # Label for producer surplus (B)

# Equilibrium point (where supply and demand intersect)
plt.scatter([equilibrium_quantity], [equilibrium_price], color='black', zorder=5)
plt.text(equilibrium_quantity, - 0.2, 'Q*', fontsize=10, ha='center')
plt.text(- 0.2, equilibrium_price, 'P*', fontsize=10, va='center')


# Set the background color to white
ax = plt.gca()
ax.set_facecolor('white')

# Hide all spines since the lines now represent the axes
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.text(-0.5, 10, 'Price', rotation=90, fontsize=12, verticalalignment='center')
plt.text(10, -0.5, 'Quantity', rotation=0, fontsize=12, verticalalignment='center')

plt.text(10, 10.5, 'Supply', rotation=0, fontsize=12, verticalalignment='center')
plt.text(10, 1.5, 'Demand', rotation=0, fontsize=12, verticalalignment='center')

# Display the plot
plt.tight_layout()
plt.savefig('Market_With_Letters.png')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create the figure and axis
fig, ax = plt.subplots(figsize=(6, 4))

# Create a blank 2x2 table with empty cells
data = [['', ''], ['', '']]
row_labels = ['Row 1', 'Row 2']
col_labels = ['Column 1', 'Column 2']

# Create a table
table = ax.table(cellText=data, 
                 rowLabels=row_labels,
                 colLabels=col_labels,
                 cellLoc='center', 
                 loc='center',
                 cellColours=[['#ffffff' for _ in range(len(col_labels))] for _ in range(len(row_labels))],
                 rowColours=['#f5f5f5', '#ffffff'],
                 colColours=['#e0e0e0'] * len(col_labels))

# Customize table appearance
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1.2, 1.2)

# Improve the appearance of cells and borders
for key, cell in table._cells.items():
    cell.set_edgecolor('black')  # Black borders for cells
    cell.set_linewidth(1)       # Border width
    cell.set_fontsize(12)       # Font size
    cell.set_text_props(weight='bold', color='black')  # Text color and weight

# Customize header appearance
for i, label in enumerate(table._colLabels):
    cell = table[(0, i)]
    cell.set_facecolor('#d0d0d0')  # Header background color
    cell.set_text_props(weight='bold', color='black')

# Hide the axes
ax.axis('off')

# Set the title (optional)
plt.title('Blank 2x2 Grid', fontsize=16, weight='bold')

# Show the plot
plt.show()
```
