### Taxes

When I walk into the store to buy a chocolate bar, if I'm willing to pay up to $\$2.00$ and the government taxes chocolate bars by $\$0.50$, I'm only able to give the seller $\$1.50$. The government may ask the seller to collect the tax. In which case I fork over $\$2$ to the seller, who then collects $\$1.50$ for themself, and gives the $\$0.50$ to the government. Or if instead the goverment asks the buyer to collect the tax. In this case, I pay the government $\$0.50$, and I can pay the seller $\$1.50$. 

In both cases, I'm paying the $\$2$, the governemnt is receiving $\$0.50$, and the seller is receiving $\$1.50$. It doesn't matter which side of the market the government taxes, the buyer's price and seller's price are the same in both cases. 

There are three ways of thinking about the tax. If it's placed on the buyer, because I'm now willing to pay $\$0.50$ less than before, my individual demand curve with the tax is that much lower than before the tax. We can then use that demand curve with the tax to find equilibrium. If the tax is palced on the seller, the seller needs to compensate for the tax by raising their costs by that $\$0.50$ to sell the same quantity. So we say the supply curve after the tax is higher by $\$0.50$. But both these methods can be confusing. 

The third way simply realizes that we have a difference between the buyer's price and the seller's price. And the difference is equal to the tax. But we still need the market to clear. So we want the quantity that intersects the supply curve at a point lower than the demand curve by that $\$0.50$. This is what we call the tax wedge. You can immagine you're sliding the tax wedge in from the left until it gets stuck between the two curves. That's the equilibrium with the tax.
