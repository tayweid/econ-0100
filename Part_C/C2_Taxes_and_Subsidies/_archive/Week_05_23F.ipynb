## Introduction to the Week

On Tuesday we will take a step back and consider how markets fit into the PPF framework we began with. Then refresh how the government gathers resources and influences behavior by imposing taxes on the sale of goods and services in a market. The major takeaways are:

- The government does not control which side of the market, consumers or producers, ends up paying the tax. These things are determined by the nature of the supply and demand curves in the market.
- The reduction is surplus value caused by a tax is generally larger than the amount of revenue raised by a tax. Minimizing this excess burden of the tax, however, would result in a tax system that many would find unfair.

We'll then break into groups to examine how governments can alter the distribution of the gains from exchange. The market divides the gains from exchange into consumer and producer surplus without any consideration of fairness or justice. It is possible for the government to redistribute the gains from exchange by setting a minimum or maximum price, but at a cost of reducing the overall levels of gain. We will look at three policies:

- A price ceiling which is designed to redistribute producer surplus to consumers. Rent control is a commonly used price ceiling.
- A price floor which is designed to redistribute consumer surplus to producers. A minimum wage law is a good example of a price floor.
- Government purchases of a good at above market prices is another way to redistribute surplus value from consumers to producers. Agricultural price supports are often done this way.

On Thursday we will use the supply and demand model to examine government trade policy. We will find that the market model suggests that

- Opening a country to trade creates winners and losers, and the winners gain more than the losers lose.
- Restricting trade by imposing import tariffs or quotas redistributes surplus value and decreases the overall amount of surplus value in the economy

### PPF Tieback

We've spent time talking about markets, the preferences of buyers and sellers, and the incentives that push markets toward equilibrium. And we found that this equilibrium gave us two things: a price and a quantity. 

At the beginning of the semester we began by framing the class in the context of the PPF. We had our two farmers who found that if they specialized and traded with each other, they could do better than they could on their own (in autarky). 

But we were left with the question of where on the PPF to live. If we fully specialized, we knew what the point on the PPF would be. But we wouldn't necessarily always fully specialize. We could partially specialize, and still be outside the autarky PPF. So that was a question.

And we proposed markets as a way to picking the point, the pair of quantities. So we built up the intuition for the demand curve, which tells us how much buyers will demand at all possible prices, which satisfies the law of demand. And we waved our hands a little, but we said something similar would work for sellers, giving us the supply curve, which we said satisfies the law of supply. 

Then we developed our intuition for the market equilibrium. Buyers and sellers have incentive to move the prices and quantities toward market clearing: the point where everything that's produced by a seller is sold to a buyer. And this equilibrium idea is what gave us a price and quantity. 

We've only ever looked at a single market. But in the background we know there must be markets for all the goods and services we're considering. So going back to the farmers, we can apply our notion of markets to tell us something about prices and quantities for spinach and carrots. 

Remember we found an exchange rate for trading between the two farmers? We said the exchange rate allowed the farmers to not just guild, but exchange with each other as separate entities while still benefitting from specialization and trade. We said this exchange rate must live between their own internal exchange rates, represented by their PPFs. We didn't decide which one was best, but said that there were a range of prices that would make the trade benefit both famers. 

What we're doing now with markets pins this exchange rate down. Using a market for spinach and a separate market for carrots, we have buyers and sellers making up the demand and supply curves, all facing the incentives that push the market toward equilibrium. The equilibrium generates prices in both markets. And these prices coordinate the decisions of the farmers by forming an exchange rate that allows the farmer to specialize acordingly. 

This is some of what's happening in the background for the farmer, who makes up part of the supply curve. If the farmer is going to produce a crop (and we'll come back to exactly how the farmer would think about it), their decision about WHICH crop comes from their own production functions and the prices in the markets. When prices are relatively high in the carrot market, both farmers will grow carrots. And the same goes for spinach. But when the prices live somewhere in the middle, between the two farmer's opportunity costs, then one will grow one crop and the other the other.

And since there are many farmers in the market with different PPFs (and hence different production functions), when working effectively, the market will arrive at a pair of prices that supply the appropriate quantities to the market once every farmer has decided their crop choices.

### Taxes

When I walk into the store to buy a chocolate bar, if I'm willing to pay up to $\$2.00$ and the government taxes chocolate bars by $\$0.50$, I'm only able to give the seller $\$1.50$. The government may ask the seller to collect the tax. In which case I fork over $\$2$ to the seller, who then collects $\$1.50$ for themself, and gives the $\$0.50$ to the government. Or if instead the goverment asks the buyer to collect the tax. In this case, I pay the government $\$0.50$, and I can pay the seller $\$1.50$. 

In both cases, I'm paying the $\$2$, the governemnt is receiving $\$0.50$, and the seller is receiving $\$1.50$. It doesn't matter which side of the market the government taxes, the buyer's price and seller's price are the same in both cases. 

There are three ways of thinking about the tax. If it's placed on the buyer, because I'm now willing to pay $\$0.50$ less than before, my individual demand curve with the tax is that much lower than before the tax. We can then use that demand curve with the tax to find equilibrium. If the tax is palced on the seller, the seller needs to compensate for the tax by raising their costs by that $\$0.50$ to sell the same quantity. So we say the supply curve after the tax is higher by $\$0.50$. But both these methods can be confusing. 

The third way simply realizes that we have a difference between the buyer's price and the seller's price. And the difference is equal to the tax. But we still need the market to clear. So we want the quantity that intersects the supply curve at a point lower than the demand curve by that $\$0.50$. This is what we call the tax wedge. You can immagine you're sliding the tax wedge in from the left until it gets stuck between the two curves. That's the equilibrium with the tax.

### Surplus Value Intuition

But lets say I walk into the store to buy a chocolate bar, I'm willing to pay $\$2$ but the seller is selling for $\$1$. This is good for me right? I don't need to pay what I would be willing to, so I've saved a dollar. This dollar, the difference between what I would be willing to pay and what I actually have to pay, is what we call Consumer Surplus. And we represent it on the graph as:
1. The area under the demand curve
2. Above the price
3. And at every quantity that's exchanged

And for sellers we have a similar notion, called Producer Surplus. This is just the difference between what a seller is willing to sell for and what they actually sell for. And we represent it on the graph in a similar way:
1. The area above the supply curve
2. Below the price
3. And at every quantity that's exchanged
